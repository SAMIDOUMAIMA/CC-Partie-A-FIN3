## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** régression linéaire (*linear regression*, moindres carrés ordinaires). **Famille :** régression supervisée. **Objectif :** prédire une cible numérique $y$ à partir de variables explicatives $x_1,\dots,x_p$.
- **Intuition :** on cherche l'hyperplan qui passe « au plus près » des observations, la proximité étant mesurée par les carrés des écarts verticaux.
- **Étapes :** données $(X,y)$ → calcul des coefficients minimisant l'erreur quadratique (solution fermée ou algorithme numérique) → prédiction $\hat y$ pour de nouvelles observations.
- **Critère :** $\hat y_i=\beta_0+\sum_{j=1}^{p}\beta_j x_{ij}$ et $\min_{\beta}\sum_{i=1}^{n}(y_i-\hat y_i)^2$, où $n$ est le nombre d'observations, $x_{ij}$ la valeur de la variable $j$ pour l'observation $i$, $\beta_0$ l'ordonnée à l'origine et $\beta_j$ les coefficients.
- **Paramètres / hyperparamètres :** les $\beta$ sont des **paramètres** appris ; la version de base n'a **pas d'hyperparamètre** (on en introduit avec Ridge, Lasso, etc.).
- **Avantages :** (1) très interprétable (un coefficient = effet d'une variable, les autres étant fixées) ; (2) rapide et stable à entraîner.
- **Limites :** (1) suppose une relation linéaire entre variables et cible ; (2) sensible aux valeurs extrêmes et à la colinéarité entre variables.

### Schéma explicatif

**Comment lire ce schéma :** Chaque point bleu est une observation ; la droite rouge est le modèle. Les segments gris sont les erreurs (résidus) : la méthode choisit la droite qui rend la somme de leurs carrés minimale.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 1
g = np.random.default_rng(0)
x = np.sort(g.uniform(0, 10, 25)); y = 2 + 0.8*x + g.normal(0, 1.2, 25)
b1, b0 = np.polyfit(x, y, 1); yh = b0 + b1*x
fig, ax = plt.subplots(figsize=(6.5, 4))
ax.scatter(x, y, c='#2563eb', label=r'observations $(x_i, y_i)$')
ax.plot(x, yh, c='#dc2626', label=r'droite ajustée $\hat{y}=\beta_0+\beta_1 x$')
ax.vlines(x, y, yh, colors='#9ca3af', lw=1, label=r'résidus $y_i-\hat{y}_i$')
ax.set_xlabel('variable explicative $x$'); ax.set_ylabel('cible $y$')
ax.set_title('Régression linéaire : on minimise la somme des carrés des segments gris'); ax.legend(fontsize=8)
plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (marketing / gestion) :** expliquer et prévoir le chiffre d'affaires mensuel d'une enseigne à partir du budget publicitaire, du prix moyen et du nombre de points de vente.
- **Cas d'usage 2 :** estimer le prix d'un logement à partir de sa surface, de son âge et de sa localisation codée.
- **Données et prérequis :** tableau de variables numériques (ou catégorielles encodées), cible continue ; plus d'observations que de variables ; variables peu corrélées entre elles ; mise à l'échelle utile si l'on compare des coefficients.
- **Peu approprié :** relation fortement non linéaire (ex. effet de saturation d'une campagne publicitaire) ou cible binaire (une probabilité hors de $[0,1]$ n'a pas de sens) ; peu fiable aussi quand les variables sont presque identiques.
- **Métriques :** MAE, RMSE, $R^2$, et analyse des résidus (structure, hétéroscédasticité) sur un jeu de test.

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** `Xr_ent` (210 observations × 5 variables) et `yr_ent` pour l'apprentissage ; `Xr_test` / `yr_test` (70 observations) pour l'évaluation. **Hyperparamètres :** aucun.
- **Sortie :** la MAE (erreur absolue moyenne) sur le test, exprimée dans l'unité de $y$.
- **Ce qu'on peut conclure :** l'ordre de grandeur de l'erreur typique de prédiction sur des données non vues. **Ce qu'on ne peut pas conclure :** la MAE seule n'est pas interprétable sans valeur de référence (par exemple prédire la moyenne de `yr_ent`) ; le jeu est **synthétique et généré par un modèle linéaire** : c'est le cas le plus favorable à cet algorithme, on ne peut donc rien en déduire sur des données réelles.
- **Résultat observé lors de notre exécution** (Python 3.13, scikit-learn 1.9.1, graine 42) : MAE ≈ 9,35. *Vérifiez ces valeurs sur votre propre exécution : elles peuvent légèrement varier selon les versions des bibliothèques.*

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** régression Ridge (*Ridge regression*, régularisation L2). **Famille :** régression supervisée. **Objectif :** prédire une cible continue en limitant l'instabilité des coefficients.
- **Intuition :** on ajoute à la régression linéaire une pénalité qui décourage les coefficients trop grands, ce qui stabilise le modèle quand les variables sont corrélées.
- **Étapes :** standardisation des variables → minimisation de l'erreur quadratique pénalisée → prédiction.
- **Critère :** $\min_{\beta}\sum_{i=1}^{n}(y_i-\hat y_i)^2+\alpha\sum_{j=1}^{p}\beta_j^2$, où $\alpha\ge 0$ est la force de pénalisation et $\beta_j$ les coefficients (l'ordonnée à l'origine n'est pas pénalisée).
- **Paramètres / hyperparamètres :** les $\beta$ sont des **paramètres** ; $\alpha$ est un **hyperparamètre** à choisir par validation croisée (jamais sur le test final).
- **Avantages :** (1) réduit la variance et gère bien la colinéarité ; (2) solution fermée, calcul rapide.
- **Limites :** (1) ne met aucun coefficient exactement à zéro, donc pas de sélection de variables ; (2) la standardisation est nécessaire car la pénalité dépend de l'échelle des variables.

### Schéma explicatif

**Comment lire ce schéma :** Chaque courbe est un coefficient en fonction de $\alpha$ (échelle logarithmique). Plus on va vers la droite, plus la pénalité est forte et plus les coefficients sont tirés vers 0, sans jamais l'atteindre exactement.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 2
from sklearn.linear_model import Ridge
g = np.random.default_rng(1)
X = g.normal(size=(60, 6)); X[:, 1] = X[:, 0] + 0.1*g.normal(size=60)   # variables 1 et 2 très corrélées
y = X @ np.array([3, 0, 1.5, 0, -2, 0.5]) + g.normal(0, 1, 60)
alphas = np.logspace(-2, 3, 40)
coefs = np.array([Ridge(alpha=a).fit(X, y).coef_ for a in alphas])
fig, ax = plt.subplots(figsize=(6.5, 4))
for j in range(6):
    ax.plot(alphas, coefs[:, j], label=rf'$\beta_{j+1}$')
ax.set_xscale('log'); ax.axhline(0, c='k', lw=.6)
ax.set_xlabel(r'pénalisation $\alpha$ (échelle log)'); ax.set_ylabel('valeur du coefficient')
ax.set_title('Ridge : les coefficients rétrécissent vers 0 quand $\\alpha$ augmente, sans jamais s\'annuler exactement')
ax.title.set_fontsize(9); ax.legend(ncol=3, fontsize=8); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance) :** modéliser le rendement d'un portefeuille à partir de nombreux indicateurs macroéconomiques fortement corrélés.
- **Cas d'usage 2 :** prédire un coût logistique à partir de dizaines de variables de trajet liées entre elles (distance, durée, carburant...).
- **Données et prérequis :** variables numériques standardisées, cible continue ; utile quand $p$ est grand ou quand les variables sont colinéaires.
- **Peu approprié :** quand on veut un modèle parcimonieux (on souhaite éliminer des variables) ou quand la relation est fortement non linéaire.
- **Métriques :** MAE, RMSE, $R^2$ en validation croisée ; courbe de validation en fonction de $\alpha$.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données que l'exemple 1, passées dans un *pipeline* `StandardScaler` → `Ridge`. **Hyperparamètre :** `alpha=1.0`, fixé arbitrairement (non réglé).
- **Sortie :** la MAE sur le jeu de test, à comparer à celle de la régression linéaire sur le **même** découpage.
- **Ce qu'on peut conclure :** si la MAE est proche de celle de l'exemple 1, la pénalisation n'apporte presque rien ici. **Ce qu'on ne peut pas conclure :** que `alpha=1.0` est un bon choix (aucune recherche d'hyperparamètre) ; qu'un écart minime entre deux MAE soit significatif (un seul découpage, données synthétiques peu colinéaires, donc peu favorables à Ridge).
- **Résultat observé lors de notre exécution** (Python 3.13, scikit-learn 1.9.1, graine 42) : MAE ≈ 9,37, quasiment identique à la régression linéaire : la pénalisation n'apporte rien de visible ici. *Vérifiez ces valeurs sur votre propre exécution : elles peuvent légèrement varier selon les versions des bibliothèques.*

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Lasso (*Least Absolute Shrinkage and Selection Operator*, régularisation L1). **Famille :** régression supervisée. **Objectif :** prédire une cible continue tout en sélectionnant automatiquement les variables utiles.
- **Intuition :** la pénalité sur la somme des valeurs absolues des coefficients force certains d'entre eux à devenir exactement nuls.
- **Étapes :** standardisation → optimisation (descente de coordonnées) de l'erreur pénalisée → prédiction avec les variables conservées.
- **Critère :** $\min_{\beta}\ \frac{1}{2n}\sum_{i=1}^{n}(y_i-\hat y_i)^2+\alpha\sum_{j=1}^{p}|\beta_j|$, avec $\alpha\ge0$ la force de pénalisation (la division par $2n$ est la convention de scikit-learn).
- **Paramètres / hyperparamètres :** $\beta$ **paramètres** ; $\alpha$ et `max_iter` (nombre maximal d'itérations de l'optimiseur) **hyperparamètres**.
- **Avantages :** (1) produit un modèle parcimonieux et lisible ; (2) utile quand $p$ est grand devant $n$.
- **Limites :** (1) instable si des variables sont très corrélées (il en choisit une un peu arbitrairement) ; (2) tend à sous-estimer les grands coefficients (biais).

### Schéma explicatif

**Comment lire ce schéma :** Chaque courbe est un coefficient en fonction de $\alpha$. Contrairement à Ridge, les courbes atteignent la valeur 0 puis y restent : la variable correspondante est éliminée du modèle.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 3
from sklearn.linear_model import Lasso
g = np.random.default_rng(1)
X = g.normal(size=(60, 6))
y = X @ np.array([3, 0, 1.5, 0, -2, 0.5]) + g.normal(0, 1, 60)
alphas = np.logspace(-2, 0.7, 40)
coefs = np.array([Lasso(alpha=a, max_iter=10000).fit(X, y).coef_ for a in alphas])
fig, ax = plt.subplots(figsize=(6.5, 4))
for j in range(6):
    ax.plot(alphas, coefs[:, j], label=rf'$\beta_{j+1}$')
ax.set_xscale('log'); ax.axhline(0, c='k', lw=.6)
ax.set_xlabel(r'pénalisation $\alpha$ (échelle log)'); ax.set_ylabel('valeur du coefficient')
ax.set_title('Lasso : des coefficients deviennent exactement nuls (sélection de variables)', fontsize=9)
ax.legend(ncol=3, fontsize=8); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (gestion du risque de crédit) :** retenir, parmi des dizaines de ratios financiers, ceux qui expliquent réellement un score de risque.
- **Cas d'usage 2 :** identifier, parmi de nombreux indicateurs de performance commerciale, les quelques-uns qui prédisent le mieux les ventes.
- **Données et prérequis :** variables numériques standardisées ; hypothèse que peu de variables ont un vrai effet.
- **Peu approprié :** quand presque toutes les variables ont un petit effet utile, ou quand des groupes de variables très corrélées doivent être conservés ensemble (préférer Ridge ou Elastic Net).
- **Métriques :** MAE, RMSE, $R^2$ ; nombre de coefficients non nuls ; stabilité de la sélection entre échantillons.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données que l'exemple 1, *pipeline* `StandardScaler` → `Lasso`. **Hyperparamètres :** `alpha=.1`, `max_iter=10000`.
- **Sortie :** le coefficient de détermination $R^2$ sur le test (1 = prédiction parfaite ; 0 = pas mieux que prédire la moyenne).
- **Ce qu'on peut conclure :** la part de variance de `yr_test` expliquée par le modèle. **Ce qu'on ne peut pas conclure :** quelles variables ont été éliminées (le code n'affiche pas les coefficients : on pourrait ajouter `modele[-1].coef_`) ; ni que `alpha=.1` est optimal. ## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** régression polynomiale (*polynomial regression*). **Famille :** régression supervisée. **Objectif :** modéliser une relation courbe tout en gardant un modèle linéaire en ses paramètres.
- **Intuition :** on ajoute aux variables d'origine leurs puissances et produits croisés, puis on ajuste une régression linéaire sur ces nouvelles variables.
- **Étapes :** standardisation → création des termes polynomiaux de degré $d$ → régression pénalisée (ici Ridge) → prédiction.
- **Critère :** $\hat y=\beta_0+\sum_j\beta_j x_j+\sum_{j\le k}\beta_{jk}x_jx_k$ pour $d=2$ ; on minimise l'erreur quadratique pénalisée de la même façon que Ridge.
- **Paramètres / hyperparamètres :** les $\beta$ sont des **paramètres** ; le degré $d$ (`degree`) et la pénalité $\alpha$ (`alpha`) sont des **hyperparamètres**.
- **Avantages :** (1) capte des courbures et des interactions simples ; (2) reste interprétable et rapide pour un petit degré.
- **Limites :** (1) le nombre de termes explose avec $p$ et $d$ ; (2) risque fort de sur-apprentissage et d'oscillations aux bords si $d$ est grand.

### Schéma explicatif

**Comment lire ce schéma :** Trois ajustements des mêmes points : degré 1 trop rigide (sous-apprentissage), degré 3 proche de la tendance, degré 12 qui épouse le bruit et oscille fortement (sur-apprentissage).

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 4
g = np.random.default_rng(2)
x = np.sort(g.uniform(0, 6, 20)); y = np.sin(x) + g.normal(0, .25, 20)
xx = np.linspace(0, 6, 300)
fig, axes = plt.subplots(1, 3, figsize=(11, 3.4), sharey=True)
for ax, d, titre in zip(axes, [1, 3, 12], ['degré 1 : sous-apprentissage', 'degré 3 : compromis', 'degré 12 : sur-apprentissage']):
    c = np.polyfit(x, y, d); ax.scatter(x, y, s=18, c='#2563eb')
    ax.plot(xx, np.polyval(c, xx), c='#dc2626'); ax.set_ylim(-2, 2); ax.set_title(titre, fontsize=9); ax.set_xlabel('$x$')
axes[0].set_ylabel('$y$'); fig.suptitle('Régression polynomiale : le degré règle la flexibilité de la courbe', fontsize=10, y=1.06); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (marketing) :** modéliser des rendements décroissants : l'effet d'une dépense publicitaire sur les ventes croît de moins en moins vite.
- **Cas d'usage 2 :** ajuster la consommation énergétique d'un bâtiment en fonction de la température extérieure (relation en courbe).
- **Données et prérequis :** peu de variables explicatives, assez d'observations par rapport au nombre de termes créés, variables standardisées.
- **Peu approprié :** beaucoup de variables (explosion combinatoire) ou besoin d'extrapoler hors du domaine observé (le polynôme diverge).
- **Métriques :** MAE, RMSE, $R^2$ sur validation ; courbe de validation selon le degré.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

- **Entrées :** mêmes données de régression que l'exemple 1. **Hyperparamètres :** `degree=2`, `alpha=10`, `include_bias=False` (le biais est déjà géré par la régression).
- **Sortie :** la MAE sur le test, à comparer aux exemples 1 et 2.
- **Ce qu'on peut conclure :** si l'erreur n'est pas meilleure que celle du modèle linéaire, les termes de degré 2 n'apportent pas d'information utile. **Ce qu'on ne peut pas conclure :** que la régression polynomiale est mauvaise en général : le jeu a été **généré par un modèle linéaire**, il n'y a donc aucune courbure à capter ici ; le degré n'a pas été réglé non plus.
- **Résultat observé lors de notre exécution** (Python 3.13, scikit-learn 1.9.1, graine 42) : MAE ≈ 10,90, moins bonne que la régression linéaire (≈ 9,35) : cohérent avec le fait que les termes de degré 2 n'ont rien à capter dans un jeu linéaire. *Vérifiez ces valeurs sur votre propre exécution : elles peuvent légèrement varier selon les versions des bibliothèques.*

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** arbre de régression (*regression tree*, CART). **Famille :** régression supervisée. **Objectif :** prédire une valeur numérique par une suite de règles « si... alors ».
- **Intuition :** on découpe récursivement l'espace des variables en zones les plus homogènes possible ; chaque zone (feuille) prédit la moyenne de ses observations d'apprentissage.
- **Étapes :** choix de la meilleure coupure (variable, seuil) → division en deux nœuds → répétition jusqu'à une condition d'arrêt → prédiction = moyenne de la feuille atteinte.
- **Critère :** à chaque nœud on minimise $\sum_{i\in G}(y_i-\bar y_G)^2+\sum_{i\in D}(y_i-\bar y_D)^2$, où $G$ et $D$ sont les groupes gauche et droit après coupure et $\bar y_G$, $\bar y_D$ leurs moyennes.
- **Paramètres / hyperparamètres :** variables et seuils de coupure = **paramètres** appris ; `max_depth` (profondeur maximale), taille minimale des feuilles = **hyperparamètres**.
- **Avantages :** (1) lisible et sans mise à l'échelle nécessaire ; (2) capte non-linéarités et interactions.
- **Limites :** (1) instable (une petite variation des données change l'arbre) et sur-apprend si trop profond ; (2) prédiction en escalier, sans extrapolation.

### Schéma explicatif

**Comment lire ce schéma :** Les traits verticaux bleus sont les coupures choisies par l'arbre ; la courbe rouge est la prédiction : constante par intervalle. Avec une profondeur 5 l'arbre suit mieux les points, mais colle aussi davantage au bruit.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 5
from sklearn.tree import DecisionTreeRegressor
g = np.random.default_rng(3)
x = np.sort(g.uniform(0, 6, 80)); y = np.sin(x) + g.normal(0, .2, 80); xx = np.linspace(0, 6, 400)
fig, axes = plt.subplots(1, 2, figsize=(10, 3.6), sharey=True)
for ax, d in zip(axes, [2, 5]):
    t = DecisionTreeRegressor(max_depth=d, random_state=0).fit(x[:, None], y)
    ax.scatter(x, y, s=12, c='#9ca3af'); ax.plot(xx, t.predict(xx[:, None]), c='#dc2626', lw=2)
    for s in sorted(t.tree_.threshold[t.tree_.feature >= 0]):
        ax.axvline(s, c='#2563eb', lw=.6, ls='--')
    ax.set_title(f'profondeur max = {d} : prédiction constante par feuille', fontsize=9); ax.set_xlabel('$x$')
axes[0].set_ylabel('$y$'); fig.suptitle('Arbre de régression : coupures (tirets bleus) et prédiction en escalier (rouge)', fontsize=10); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (commerce / logistique) :** prévoir le délai de livraison selon la distance, le mode de transport et le type de produit, avec des règles lisibles par les opérationnels.
- **Cas d'usage 2 :** estimer la valeur d'un véhicule d'occasion selon l'âge et le kilométrage.
- **Données et prérequis :** variables numériques ou catégorielles encodées ; scikit-learn ne gère pas directement les catégories textuelles.
- **Peu approprié :** relation lisse et linéaire, ou besoin d'extrapoler au-delà des valeurs vues (l'arbre reste plat) ; modèle seul peu stable.
- **Métriques :** MAE, RMSE, $R^2$ ; courbe de validation selon la profondeur.

```python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** données de régression de l'exemple 1. **Hyperparamètres :** `max_depth=5`, `random_state=42` (pour départager des coupures équivalentes).
- **Sortie :** la MAE du test.
- **Ce qu'on peut conclure :** le niveau d'erreur d'un arbre de profondeur fixée sur ce découpage. **Ce qu'on ne peut pas conclure :** que la profondeur 5 est la bonne (non réglée) ; que l'arbre est meilleur ou moins bon que la régression linéaire sans comparer les métriques sur le même test ; sur des données générées linéairement, un modèle en escalier est naturellement désavantagé.
- **Résultat observé lors de notre exécution** (Python 3.13, scikit-learn 1.9.1, graine 42) : MAE ≈ 37,72, très supérieure à celle de la régression linéaire (≈ 9,35) : la prédiction en escalier est mal adaptée à une relation linéaire. *Vérifiez ces valeurs sur votre propre exécution : elles peuvent légèrement varier selon les versions des bibliothèques.*

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** forêt aléatoire pour la régression (*Random Forest regressor*). **Famille :** régression supervisée (méthode d'ensemble par *bagging*). **Objectif :** prédire une valeur numérique plus stable qu'un arbre unique.
- **Intuition :** on entraîne de nombreux arbres sur des échantillons différents puis on moyenne leurs prédictions ; les erreurs individuelles se compensent en partie.
- **Étapes :** pour chaque arbre, tirage d'un échantillon bootstrap (avec remise) → construction de l'arbre (à chaque coupure, `max_features` peut limiter les variables examinées ; par défaut, scikit-learn les examine toutes en régression) → moyenne des prédictions des arbres.
- **Critère :** $\hat y(x)=\frac{1}{B}\sum_{b=1}^{B}T_b(x)$, où $B$ est le nombre d'arbres et $T_b(x)$ la prédiction du $b$-ième arbre.
- **Paramètres / hyperparamètres :** les arbres sont les **paramètres appris** ; `n_estimators` ($B$), `max_depth`, `max_features`, `min_samples_leaf` sont des **hyperparamètres**.
- **Avantages :** (1) robuste, peu sensible au réglage et aux valeurs extrêmes ; (2) fournit une importance des variables.
- **Limites :** (1) moins interprétable qu'un arbre ; (2) ne sait pas extrapoler et peut être coûteuse en mémoire.

### Schéma explicatif

**Comment lire ce schéma :** Les lignes bleues claires sont les prédictions de 15 arbres entraînés sur des échantillons bootstrap ; la ligne rouge épaisse, leur moyenne, est plus régulière que n'importe quel arbre pris seul.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 6
from sklearn.tree import DecisionTreeRegressor
g = np.random.default_rng(4)
x = np.sort(g.uniform(0, 6, 100)); y = np.sin(x) + g.normal(0, .3, 100); xx = np.linspace(0, 6, 300)
preds = []
fig, ax = plt.subplots(figsize=(7, 4))
for b in range(15):
    idx = g.integers(0, len(x), len(x))                       # échantillon bootstrap
    t = DecisionTreeRegressor(max_depth=5, random_state=b).fit(x[idx, None], y[idx])
    p = t.predict(xx[:, None]); preds.append(p); ax.plot(xx, p, c='#93c5fd', lw=.8, label='arbres individuels (bootstrap)' if b == 0 else None)
ax.plot(xx, np.mean(preds, axis=0), c='#dc2626', lw=2.5, label='moyenne de la forêt')
ax.scatter(x, y, s=8, c='#6b7280'); ax.set_xlabel('$x$'); ax.set_ylabel('$y$'); ax.legend(fontsize=8)
ax.set_title('Forêt aléatoire : la moyenne lisse la variabilité des arbres', fontsize=10); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance) :** prévoir le montant de sinistres ou de pertes à partir de nombreuses caractéristiques hétérogènes.
- **Cas d'usage 2 :** estimer la demande d'un produit (logistique) à partir de la saison, du prix et de promotions.
- **Données et prérequis :** tableau de variables mixtes (catégorielles encodées) ; pas de mise à l'échelle nécessaire ; mémoire suffisante pour stocker les arbres.
- **Peu approprié :** séries temporelles avec tendance à extrapoler, ou contexte exigeant des règles totalement explicables.
- **Métriques :** MAE, RMSE, $R^2$ ; erreur *out-of-bag* ; importance des variables.

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

- **Sortie :** la MAE du test.
- **Ce qu'on peut conclure :** l'erreur moyenne d'une forêt de 80 arbres sur ce test. **Ce qu'on ne peut pas conclure :** que le nombre d'arbres est optimal ni que la forêt bat les modèles linéaires : le jeu est linéaire par construction, donc favorable aux modèles linéaires. Un seul découpage et une seule graine ne permettent pas de mesurer la variabilité du résultat.

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** boosting de gradient pour la régression (*Gradient Boosting*, ici `HistGradientBoostingRegressor`). **Famille :** régression supervisée (ensemble séquentiel). **Objectif :** réduire l'erreur en corrigeant progressivement les erreurs d'un modèle initial.
- **Intuition :** on construit des petits arbres l'un après l'autre ; chacun apprend à prédire les erreurs (résidus) laissées par la somme des précédents.
- **Étapes :** prédiction initiale (moyenne) → calcul des résidus → ajustement d'un arbre sur ces résidus → ajout de sa contribution, réduite par un taux d'apprentissage → répétition.
- **Critère :** $F_m(x)=F_{m-1}(x)+\eta\,h_m(x)$, où $F_m$ est le modèle après $m$ arbres, $h_m$ le $m$-ième arbre entraîné sur les résidus et $\eta$ le taux d'apprentissage ; la variante « Hist » discrétise les variables en classes (histogrammes) pour accélérer le calcul.
- **Paramètres / hyperparamètres :** les arbres sont les **paramètres appris** ; `max_iter` (nombre d'arbres), `learning_rate` ($\eta$), `max_depth` sont des **hyperparamètres**.
- **Avantages :** (1) souvent très performant sur données tabulaires ; (2) gère valeurs manquantes et non-linéarités.
- **Limites :** (1) sensible au réglage (trop d'arbres → sur-apprentissage) ; (2) entraînement séquentiel et moins interprétable.

### Schéma explicatif

**Comment lire ce schéma :** De gauche à droite, la prédiction rouge après 1, 5 puis 60 arbres : elle part d'une forme grossière et se rapproche de la tendance des points ; avec trop d'arbres elle commencerait à suivre le bruit.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 7
from sklearn.ensemble import GradientBoostingRegressor
g = np.random.default_rng(5)
x = np.sort(g.uniform(0, 6, 120)); y = np.sin(x) + g.normal(0, .25, 120); xx = np.linspace(0, 6, 300)
m = GradientBoostingRegressor(n_estimators=60, max_depth=2, learning_rate=.3, random_state=0).fit(x[:, None], y)
etapes = {1: None, 5: None, 60: None}
for k, p in enumerate(m.staged_predict(xx[:, None]), 1):
    if k in etapes: etapes[k] = p
fig, axes = plt.subplots(1, 3, figsize=(11, 3.4), sharey=True)
for ax, (k, p) in zip(axes, etapes.items()):
    ax.scatter(x, y, s=8, c='#9ca3af'); ax.plot(xx, p, c='#dc2626', lw=2); ax.set_title(f'après {k} arbre(s)', fontsize=9); ax.set_xlabel('$x$')
axes[0].set_ylabel('$y$'); fig.suptitle('Gradient boosting : chaque arbre corrige les erreurs résiduelles du modèle précédent', fontsize=10); plt.show()
```

- **Données et prérequis :** tableau de variables, de préférence plusieurs milliers d'observations ; validation soignée (arrêt précoce, validation croisée).
- **Peu approprié :** très petits échantillons (risque de sur-apprentissage) ou contexte où chaque décision doit être expliquée simplement.
- **Métriques :** MAE, RMSE, $R^2$ ; courbes d'apprentissage selon le nombre d'itérations.

```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele=HistGradientBoostingRegressor(max_iter=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

- **Sortie :** la MAE du test.
- **Ce qu'on peut conclure :** l'erreur de ce modèle précis sur ce test. **Ce qu'on ne peut pas conclure :** que le boosting est supérieur ou inférieur aux autres régressions (même remarque que pour la forêt : données linéaires, un seul découpage) ; l'effet du nombre d'itérations n'est pas étudié (aucune courbe de validation).

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** régression logistique (*logistic regression*). **Famille :** classification supervisée. **Objectif :** estimer la probabilité d'appartenir à une classe et décider selon un seuil.
- **Intuition :** on calcule un score linéaire des variables, qu'une fonction sigmoïde transforme en probabilité comprise entre 0 et 1.
- **Étapes :** standardisation → estimation des coefficients par maximum de vraisemblance (avec pénalité L2 par défaut) → probabilité $\hat p$ → classe 1 si $\hat p\ge0{,}5$ (seuil modifiable).
- **Critère :** $P(y=1\mid x)=\dfrac{1}{1+e^{-(\beta_0+\sum_j\beta_jx_j)}}$ ; on minimise la perte logistique $-\sum_i[y_i\log\hat p_i+(1-y_i)\log(1-\hat p_i)]$ (plus une pénalité), où $\hat p_i$ est la probabilité prédite pour l'observation $i$.
- **Paramètres / hyperparamètres :** les $\beta$ sont des **paramètres** ; `C` (inverse de la force de régularisation) et `max_iter` sont des **hyperparamètres**.
- **Avantages :** (1) sortie probabiliste et coefficients interprétables (rapports de cotes) ; (2) rapide et solide comme modèle de référence.
- **Limites :** (1) frontière de décision linéaire ; (2) sensible au déséquilibre de classes et à la colinéarité.

### Schéma explicatif

**Comment lire ce schéma :** Les points rouges et bleus sont les observations des classes 1 et 0 ; la courbe noire est la probabilité estimée de la classe 1 en fonction de $x$. Le seuil 0,5 (tirets) fixe la frontière de décision.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 8
from sklearn.linear_model import LogisticRegression
g = np.random.default_rng(6)
x = np.concatenate([g.normal(-1.2, 1, 40), g.normal(1.2, 1, 40)]); y = np.r_[np.zeros(40), np.ones(40)]
m = LogisticRegression().fit(x[:, None], y); xx = np.linspace(-5, 5, 300)
p = m.predict_proba(xx[:, None])[:, 1]
fig, ax = plt.subplots(figsize=(6.5, 4))
ax.scatter(x, y, c=y, cmap='coolwarm', s=18, label='observations (classe 0 ou 1)')
ax.plot(xx, p, c='k', label=r'$P(y=1\mid x)=1/(1+e^{-(\beta_0+\beta_1 x)})$')
ax.axhline(.5, c='gray', ls='--', lw=.8); ax.axvline(-m.intercept_[0]/m.coef_[0, 0], c='gray', ls=':', lw=.8)
ax.text(-4.8, .53, 'seuil 0,5', fontsize=8); ax.set_xlabel('score / variable $x$'); ax.set_ylabel('probabilité')
ax.set_title('Régression logistique : une sigmoïde transforme un score linéaire en probabilité', fontsize=9); ax.legend(fontsize=8, loc='center right'); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance) :** estimer la probabilité de défaut d'un emprunteur (*credit scoring*).
- **Cas d'usage 2 :** prédire si un client va répondre à une campagne de marketing.
- **Données et prérequis :** variables numériques standardisées, cible binaire (ou multiclasse) ; classes pas trop déséquilibrées ou pondérées.
- **Peu approprié :** frontière fortement non linéaire (ex. classes imbriquées en cercles) sans création de variables.
- **Métriques :** F1, précision, rappel, ROC-AUC, PR-AUC si classe rare ; calibration des probabilités.

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** k plus proches voisins (*k-nearest neighbors*, KNN). **Famille :** classification supervisée (méthode à base d'instances). **Objectif :** classer une observation d'après ses voisins les plus proches.
- **Intuition :** « dis-moi qui sont tes voisins, je te dirai qui tu es » : la classe est celle qui domine parmi les $k$ points d'entraînement les plus proches.
- **Étapes :** standardisation (indispensable) → stockage des données d'entraînement → pour un nouveau point, calcul des distances → sélection des $k$ voisins → vote majoritaire.
- **Critère :** $\hat y(x)=\arg\max_c\sum_{i\in V_k(x)}\mathbf{1}[y_i=c]$, où $V_k(x)$ est l'ensemble des $k$ voisins de $x$, $c$ une classe et $\mathbf{1}[\cdot]$ vaut 1 si la condition est vraie, 0 sinon ; distance usuelle : euclidienne.
- **Paramètres / hyperparamètres :** il n'y a pas de paramètres appris (les données **sont** le modèle) ; $k$ (`n_neighbors`), la métrique de distance et le mode de pondération sont des **hyperparamètres**.
- **Avantages :** (1) très simple, sans phase d'entraînement ; (2) frontière non linéaire flexible.
- **Limites :** (1) prédiction lente et gourmande en mémoire sur de gros jeux ; (2) dégradé en grande dimension et sensible à l'échelle des variables.

### Schéma explicatif

**Comment lire ce schéma :** L'étoile jaune est le point à classer ; les cercles noirs entourent ses 5 plus proches voisins, délimités par le cercle en tirets. La classe majoritaire parmi eux décide de la prédiction.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

- **Peu approprié :** très grandes bases (coût de calcul par prédiction) ou données de haute dimension où les distances perdent leur sens.
- **Métriques :** F1, rappel, précision, ROC-AUC ; choix de $k$ par validation croisée.

```python
from sklearn.neighbors import KNeighborsClassifier
modele=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Bayes naïf gaussien (*Gaussian Naive Bayes*). **Famille :** classification supervisée probabiliste. **Objectif :** classer en appliquant le théorème de Bayes avec une hypothèse d'indépendance entre variables.
- **Intuition :** pour chaque classe, on modélise chaque variable par une loi normale ; la classe prédite est celle qui rend l'observation la plus plausible compte tenu de sa fréquence.
- **Étapes :** estimation de la proportion de chaque classe et, par classe, de la moyenne et de l'écart-type de chaque variable → calcul de la probabilité a posteriori → classe de probabilité maximale.
- **Critère :** $P(c\mid x)\propto P(c)\prod_{j=1}^{p}\mathcal N(x_j;\mu_{cj},\sigma_{cj}^2)$, où $P(c)$ est la probabilité a priori de la classe $c$, $\mu_{cj}$ et $\sigma_{cj}$ la moyenne et l'écart-type de la variable $j$ dans la classe $c$.
- **Paramètres / hyperparamètres :** $P(c)$, $\mu_{cj}$ et $\sigma_{cj}$ sont les **paramètres** ; `var_smoothing` (stabilité numérique des variances) est le principal **hyperparamètre**.
- **Avantages :** (1) très rapide, efficace avec peu de données ; (2) probabilités simples et robustes à l'ajout de variables.
- **Limites :** (1) l'indépendance conditionnelle est rarement vraie ; (2) les probabilités produites sont souvent mal calibrées (trop extrêmes).

### Schéma explicatif

**Comment lire ce schéma :** Les deux courbes sont les densités de $x$ pondérées par la fréquence de chaque classe. La zone colorée en dessous de la courbe la plus haute montre quelle classe est prédite en chaque valeur de $x$.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 10
xx = np.linspace(-5, 8, 400)
def gauss(x, m, s): return np.exp(-(x-m)**2/(2*s**2))/(s*np.sqrt(2*np.pi))
f0, f1 = gauss(xx, 0, 1.2)*.6, gauss(xx, 3, 1.5)*.4            # P(x|classe) × P(classe)
post1 = f1/(f0+f1)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(xx, f0, c='#2563eb', label=r'$P(x\mid y=0)\,P(y=0)$'); ax.plot(xx, f1, c='#dc2626', label=r'$P(x\mid y=1)\,P(y=1)$')
ax.fill_between(xx, 0, np.maximum(f0, f1), where=post1 > .5, color='#dc2626', alpha=.12); ax.fill_between(xx, 0, np.maximum(f0, f1), where=post1 <= .5, color='#2563eb', alpha=.12)
ax.set_xlabel('variable $x$ (une seule dimension pour simplifier)'); ax.set_ylabel('densité pondérée')
ax.set_title('Naive Bayes gaussien : la classe prédite est celle dont la courbe est la plus haute', fontsize=9); ax.legend(fontsize=8); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (marketing / gestion) :** filtrage de messages ou classement rapide de demandes clients à partir de nombreux indicateurs.
- **Cas d'usage 2 :** diagnostic rapide d'anomalies de production à partir de mesures de capteurs.
- **Données et prérequis :** variables numériques à peu près gaussiennes par classe ; peu d'observations suffisent.
- **Peu approprié :** variables fortement dépendantes entre elles, ou quand on a besoin de probabilités bien calibrées sans recalibrage.
- **Métriques :** F1, précision, rappel, ROC-AUC, log-loss / calibration.

```python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

- **Entrées :** données de classification de l'exemple 8 (sans standardisation, inutile pour ce modèle). **Hyperparamètres :** valeurs par défaut.
- **Sortie :** le F1 sur le test.
- **Ce qu'on peut conclure :** la qualité de classification avec l'hypothèse d'indépendance sur ces données. **Ce qu'on ne peut pas conclure :** si l'hypothèse gaussienne est adaptée (non testée) ; que le modèle est bien calibré. Ici l'hypothèse d'indépendance conditionnelle des variables n'est pas vérifiée : le résultat montre seulement ce que donne le modèle malgré elle.

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** arbre de décision (*decision tree classifier*, CART). **Famille :** classification supervisée. **Objectif :** classer par une suite de questions simples sur les variables.
- **Intuition :** chaque nœud pose une question du type « $x_j\le s$ ? » qui sépare au mieux les classes ; on descend jusqu'à une feuille qui donne la classe majoritaire.
- **Étapes :** pour chaque nœud, recherche de la variable et du seuil qui réduisent le plus l'impureté → division → arrêt selon la profondeur ou la taille des feuilles → prédiction.
- **Critère :** impureté de Gini $G=1-\sum_{c}p_c^{2}$, où $p_c$ est la proportion de la classe $c$ dans le nœud ; on choisit la coupure qui diminue le plus $G$ pondéré par la taille des deux enfants.
- **Paramètres / hyperparamètres :** variables, seuils et feuilles sont les **paramètres** appris ; `max_depth`, `min_samples_leaf`, `criterion` sont des **hyperparamètres**.
- **Avantages :** (1) règles lisibles et visualisables ; (2) pas de mise à l'échelle, gère les interactions.
- **Limites :** (1) instable et sur-apprend si l'arbre est profond ; (2) frontières en « escaliers » parallèles aux axes.

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Intuition :** plusieurs arbres légèrement différents votent ; l'agrégation compense leurs erreurs individuelles.
- **Étapes :** pour chaque arbre, échantillon bootstrap et sous-ensemble aléatoire de variables à chaque coupure → entraînement → vote majoritaire (ou moyenne des probabilités) des arbres.

## 13. SVM

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** machine à vecteurs de support (*Support Vector Machine*, SVM ; ici `SVC` à noyau RBF). **Famille :** classification supervisée. **Objectif :** trouver la frontière qui sépare les classes avec la plus grande marge.
- **Intuition :** parmi toutes les frontières possibles, on choisit celle la plus éloignée des points les plus proches de chaque classe (les vecteurs de support) ; un noyau permet des frontières courbes.
- **Étapes :** standardisation → résolution d'un problème d'optimisation convexe → conservation des vecteurs de support → prédiction par le signe de la fonction de décision.
- **Critère :** $\min_{w,b,\xi}\ \frac12\|w\|^2+C\sum_i\xi_i$ sous contrainte $y_i(w\cdot\phi(x_i)+b)\ge1-\xi_i$, où $w$ et $b$ définissent l'hyperplan, $\phi$ la transformation du noyau, $\xi_i\ge0$ les violations de marge tolérées et $C$ le compromis marge/erreurs ; noyau RBF : $K(x,x')=\exp(-\gamma\|x-x'\|^2)$.
- **Paramètres / hyperparamètres :** vecteurs de support et coefficients (duaux) sont les **paramètres** ; `C`, `kernel`, `gamma` sont des **hyperparamètres**.
- **Avantages :** (1) efficace en dimension élevée et frontières non linéaires ; (2) résultat dépendant seulement des vecteurs de support.
- **Limites :** (1) coût élevé sur grands jeux ; (2) sensible au réglage de `C` et `gamma` et peu interprétable.

### Schéma explicatif

**Comment lire ce schéma :** Le schéma utilise un noyau linéaire pour rester lisible : le trait plein est la frontière, les tirets sont les bords de la marge, et les points entourés (vecteurs de support) sont les seuls qui la déterminent. Le code du notebook utilise, lui, un noyau RBF.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 13
from sklearn.datasets import make_blobs
from sklearn.svm import SVC
X, y = make_blobs(n_samples=60, centers=2, cluster_std=1.0, random_state=3)
m = SVC(kernel='linear', C=1.0).fit(X, y)
gx, gy = np.meshgrid(np.linspace(X[:, 0].min()-1, X[:, 0].max()+1, 300), np.linspace(X[:, 1].min()-1, X[:, 1].max()+1, 300))
Z = m.decision_function(np.c_[gx.ravel(), gy.ravel()]).reshape(gx.shape)
fig, ax = plt.subplots(figsize=(6, 4.6))
ax.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', s=25)
ax.contour(gx, gy, Z, levels=[-1, 0, 1], colors='k', linestyles=['--', '-', '--'])
ax.scatter(*m.support_vectors_.T, s=120, facecolors='none', edgecolors='k', lw=1.4, label='vecteurs de support')
ax.set_xlabel('variable 1'); ax.set_ylabel('variable 2'); ax.legend(fontsize=8)
ax.set_title('SVM : hyperplan (trait plein) et marges (tirets) ; seuls les vecteurs de support le déterminent', fontsize=8.5); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance) :** classer des opérations ou des clients quand les variables sont nombreuses et les observations modestes.
- **Cas d'usage 2 :** classification de documents ou de signaux de capteurs en dimension élevée.
- **Données et prérequis :** variables numériques standardisées ; jeu de taille petite à moyenne (quelques dizaines de milliers d'observations au plus).
- **Peu approprié :** très gros volumes ou besoin de probabilités bien calibrées (non fournies nativement) ; besoin d'explication simple.
- **Métriques :** F1, précision, rappel, ROC-AUC (avec la fonction de décision) ; réglage de `C` et `gamma` par validation croisée.

```python
from sklearn.svm import SVC
modele=make_pipeline(StandardScaler(),SVC(kernel='rbf',C=1.0,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** données de classification de l'exemple 8, standardisées dans le *pipeline* (indispensable pour un noyau RBF). **Hyperparamètres :** `kernel='rbf'`, `C=1.0`, `gamma` par défaut (`'scale'`).
- **Sortie :** le F1 du test.

## 14. Gradient Boosting — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** boosting de gradient pour la classification (*Gradient Boosting classifier*, ici `HistGradientBoostingClassifier`). **Famille :** classification supervisée (ensemble séquentiel). **Objectif :** classer en cumulant de petits arbres qui corrigent les erreurs précédentes.
- **Intuition :** chaque nouvel arbre est entraîné sur la direction qui réduit le plus la perte logistique des prédictions actuelles.
- **Étapes :** score initial (log-cotes de la proportion de la classe 1) → calcul du gradient de la perte → ajustement d'un arbre sur ce gradient → mise à jour du score → conversion en probabilité par sigmoïde.
- **Critère :** $F_m(x)=F_{m-1}(x)+\eta\,h_m(x)$ et $P(y=1\mid x)=1/(1+e^{-F_M(x)})$, où $F_m$ est le score après $m$ arbres, $h_m$ l'arbre entraîné sur le gradient de la perte logistique et $\eta$ le taux d'apprentissage.
- **Paramètres / hyperparamètres :** arbres = **paramètres** appris ; `max_iter`, `learning_rate`, `max_depth`, `l2_regularization` = **hyperparamètres**.
- **Avantages :** (1) souvent parmi les meilleurs modèles sur données tabulaires ; (2) gère les valeurs manquantes (variante Hist).
- **Limites :** (1) sensible au réglage et au sur-apprentissage ; (2) peu interprétable et entraînement séquentiel.

### Schéma explicatif

**Comment lire ce schéma :** Le fond coloré représente la probabilité de la classe rouge prédite par le modèle. Après 1 arbre la frontière est grossière ; après 10 puis 100 arbres elle devient plus fine et finit par épouser le bruit, signe de sur-apprentissage possible.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 14
from sklearn.datasets import make_moons
from sklearn.ensemble import GradientBoostingClassifier
X, y = make_moons(200, noise=.25, random_state=0)
m = GradientBoostingClassifier(n_estimators=100, max_depth=2, learning_rate=.3, random_state=0).fit(X, y)
gx, gy = np.meshgrid(np.linspace(-2, 3, 250), np.linspace(-1.5, 2, 250)); grille = np.c_[gx.ravel(), gy.ravel()]
etapes = {1: None, 10: None, 100: None}
for k, Z in enumerate(m.staged_predict_proba(grille), 1):
    if k in etapes: etapes[k] = Z[:, 1].reshape(gx.shape)
fig, axes = plt.subplots(1, 3, figsize=(11, 3.4), sharey=True)
for ax, (k, Z) in zip(axes, etapes.items()):
    ax.contourf(gx, gy, Z, levels=np.linspace(0, 1, 11), alpha=.35, cmap='coolwarm'); ax.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', s=10, ec='k', lw=.2)
    ax.set_title(f'après {k} arbre(s)', fontsize=9)
fig.suptitle('Gradient boosting (classification) : la probabilité prédite s\'affine arbre après arbre', fontsize=10); plt.show()
```

- **Peu approprié :** petits échantillons très bruités ou contrainte forte d'explicabilité.
- **Métriques :** F1, précision, rappel, ROC-AUC, PR-AUC si classe rare, log-loss.

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** AdaBoost (*Adaptive Boosting*). **Famille :** classification supervisée (ensemble séquentiel). **Objectif :** combiner de nombreux classifieurs faibles (par défaut des souches, arbres à une coupure) en un classifieur fort.
- **Intuition :** à chaque tour, on augmente le poids des observations mal classées pour que le prochain classifieur s'y concentre ; les classifieurs les plus précis ont le plus d'influence dans le vote final.
- **Étapes :** poids égaux → entraînement d'un classifieur faible pondéré → calcul de son erreur pondérée → mise à jour des poids → répétition → vote pondéré.
- **Critère :** $\alpha_t=\frac12\ln\frac{1-\varepsilon_t}{\varepsilon_t}$ et $H(x)=\mathrm{sign}\big(\sum_t\alpha_t h_t(x)\big)$, où $\varepsilon_t$ est l'erreur pondérée du classifieur $h_t$ au tour $t$ et $\alpha_t$ son poids dans le vote (formule du cas binaire ; scikit-learn utilise sa variante multiclasse SAMME).
- **Paramètres / hyperparamètres :** classifieurs faibles et $\alpha_t$ = **paramètres** appris ; `n_estimators`, `learning_rate`, type du classifieur de base = **hyperparamètres**.
- **Avantages :** (1) simple et efficace avec des modèles faibles ; (2) peu d'hyperparamètres.
- **Limites :** (1) sensible aux valeurs aberrantes et au bruit d'étiquettes (leur poids explose) ; (2) séquentiel donc peu parallélisable.

### Schéma explicatif

**Comment lire ce schéma :** La taille d'un point est proportionnelle à son poids. Au tour 1 tous les points ont le même poids ; les points mal classés par la première coupure grossissent au tour 2, et ainsi de suite. Chaque titre donne l'erreur pondérée et le poids du classifieur.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 15
from sklearn.datasets import make_moons
from sklearn.tree import DecisionTreeClassifier
X, y = make_moons(70, noise=.25, random_state=1); w = np.ones(len(y))/len(y)
gx, gy = np.meshgrid(np.linspace(-2, 3, 200), np.linspace(-1.5, 2, 200)); grille = np.c_[gx.ravel(), gy.ravel()]
fig, axes = plt.subplots(1, 3, figsize=(11.5, 3.6), sharey=True)
for t, ax in enumerate(axes):
    s = DecisionTreeClassifier(max_depth=1, random_state=0).fit(X, y, sample_weight=w); p = s.predict(X)
    ax.contourf(gx, gy, s.predict(grille).reshape(gx.shape), alpha=.2, cmap='coolwarm')
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', s=w*len(y)*30, ec='k', lw=.3)
    err = w[p != y].sum(); alpha = .5*np.log((1-err)/err)
    ax.set_title(f'tour {t+1} : erreur pondérée = {err:.2f}, poids du stump = {alpha:.2f}', fontsize=8.5)
    w = w*np.exp(alpha*np.where(p != y, 1, -1)); w /= w.sum()                 # on augmente le poids des points mal classés
fig.suptitle('AdaBoost : la taille des points = leur poids ; les erreurs du tour t pèsent plus au tour t+1', fontsize=10); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (gestion / marketing) :** classer des prospects en « à contacter / à ignorer » à partir de quelques indicateurs simples.
- **Cas d'usage 2 :** détection de visages historique (algorithme de Viola-Jones, fondé sur un boosting de classifieurs simples).
- **Données et prérequis :** données propres, étiquettes fiables ; classifieurs de base simples.
- **Peu approprié :** données avec beaucoup d'étiquettes erronées ou de valeurs extrêmes, car l'algorithme les sur-pondère.
- **Métriques :** F1, précision, rappel, ROC-AUC ; erreur d'apprentissage selon le nombre d'itérations.

```python
from sklearn.ensemble import AdaBoostClassifier
modele=AdaBoostClassifier(n_estimators=70,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** données de classification de l'exemple 8. **Hyperparamètres :** `n_estimators=70`, `random_state=42`, classifieur de base par défaut (souche de profondeur 1).
- **Sortie :** le F1 du test.
- **Ce qu'on peut conclure :** la performance de 70 souches pondérées sur ce découpage. **Ce qu'on ne peut pas conclure :** l'influence du taux d'apprentissage (non modifié) ; une supériorité sur les autres méthodes d'ensemble sans validation croisée.
- **Résultat observé lors de notre exécution** (Python 3.13, scikit-learn 1.9.1, graine 42) : F1 ≈ 0,827. *Vérifiez ces valeurs sur votre propre exécution : elles peuvent légèrement varier selon les versions des bibliothèques.*

## 16. XGBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** XGBoost (*eXtreme Gradient Boosting*). **Famille :** classification (et régression) supervisée, boosting de gradient optimisé. **Objectif :** obtenir un boosting d'arbres rapide, régularisé et performant.
- **Intuition :** comme le boosting de gradient, mais la perte est approchée à l'ordre 2 (gradient et hessien) et la complexité des arbres est explicitement pénalisée.
- **Étapes :** pour chaque arbre, calcul de $g_i$ et $h_i$ → recherche des meilleures coupures par un gain exact ou par histogramme → ajout de l'arbre réduit par $\eta$ → mise à jour des prédictions.
- **Critère :** $\mathcal L=\sum_i\ell(y_i,\hat y_i)+\sum_k\Omega(f_k)$ avec $\Omega(f)=\gamma T+\frac12\lambda\sum_{j=1}^{T}w_j^2$, où $\ell$ est la perte, $f_k$ le $k$-ième arbre, $T$ son nombre de feuilles, $w_j$ le score de la feuille $j$, $\gamma$ et $\lambda$ des pénalités.
- **Paramètres / hyperparamètres :** arbres et scores de feuilles = **paramètres** ; `n_estimators`, `max_depth`, `learning_rate`, `subsample`, `reg_lambda`, `reg_alpha` = **hyperparamètres**.
- **Avantages :** (1) très bonnes performances sur données tabulaires, rapide et parallélisé ; (2) régularisation intégrée et gestion des valeurs manquantes.
- **Limites :** (1) nombreux hyperparamètres à régler ; (2) moins interprétable et exige la bibliothèque `xgboost` (hors scikit-learn).

### Schéma explicatif

**Comment lire ce schéma :** Lecture de gauche à droite : les données alimentent l'arbre 1 ; l'écart entre prédiction et réalité est résumé par $g_i$ et $h_i$ ; l'arbre 2 est entraîné sur eux, et ainsi de suite jusqu'à l'arbre $K$. La prédiction finale additionne les contributions de tous les arbres. Ce schéma est un diagramme de flux (pas de frontière de décision).

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

- **Données et prérequis :** données tabulaires, installation de `xgboost`, validation soignée (arrêt précoce).
- **Peu approprié :** données non structurées brutes (images, textes) ou très petit jeu sans réglage prudent.
- **Métriques :** F1, précision, rappel, ROC-AUC, PR-AUC, log-loss ; courbes d'apprentissage.

## 17. K-means

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** K-means (*K-means clustering*, K-moyennes). **Famille :** clustering (apprentissage non supervisé). **Objectif :** partitionner des observations en $K$ groupes compacts.
- **Intuition :** chaque groupe est représenté par son centre (centroïde) ; on rattache chaque point au centre le plus proche puis on recalcule les centres, jusqu'à stabilité.
- **Étapes :** choix de $K$ → initialisation des centroïdes (ici *k-means++*, avec `n_init=10` essais) → assignation → recalcul des centroïdes → arrêt quand les assignations ne changent plus.
- **Critère :** $\min_{C}\sum_{k=1}^{K}\sum_{x_i\in C_k}\|x_i-\mu_k\|^2$, où $C_k$ est le groupe $k$ et $\mu_k$ son centroïde (moyenne de ses points) ; c'est l'inertie.
- **Paramètres / hyperparamètres :** les centroïdes $\mu_k$ sont les **paramètres** appris ; $K$ (`n_clusters`), `n_init`, `init` sont des **hyperparamètres**.
- **Avantages :** (1) simple, rapide, passe à grande échelle ; (2) résultat facile à interpréter (groupes représentés par un profil moyen).
- **Limites :** (1) $K$ doit être fixé à l'avance et le résultat dépend de l'initialisation ; (2) suppose des groupes compacts, de taille et de forme comparables, et est sensible aux valeurs extrêmes.

### Schéma explicatif

**Comment lire ce schéma :** Panneau 1 : trois centroïdes (croix rouges) au départ. Panneaux suivants : les points prennent la couleur du centroïde le plus proche, puis les croix se déplacent au centre de leur groupe. Après quelques itérations, les croix ne bougent presque plus.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 17
from sklearn.datasets import make_blobs
X, yb = make_blobs(n_samples=150, centers=3, cluster_std=.9, random_state=4)
c = np.array([X[yb == k][np.argmax(X[yb == k][:, 0])] for k in range(3)])      # centroïdes initiaux : un point en bord de chaque nuage
fig, axes = plt.subplots(1, 4, figsize=(14, 3.5), sharey=True, sharex=True)
for it, ax in enumerate(axes):
    if it == 0:
        lab = np.zeros(len(X), int)
        ax.scatter(X[:, 0], X[:, 1], s=10, c='#9ca3af'); ax.set_title('départ : 3 centroïdes initiaux', fontsize=9)
    else:
        lab = ((X[:, None, :]-c[None])**2).sum(2).argmin(1)
        ax.scatter(X[:, 0], X[:, 1], s=10, c=lab, cmap='viridis'); ax.set_title(f'itération {it} : assignation', fontsize=9)
    ax.scatter(*c.T, marker='X', s=140, c='red', ec='k', label='centroïdes')
    if it > 0: c = np.array([X[lab == k].mean(0) if (lab == k).any() else c[k] for k in range(3)])    # recalcul des centroïdes
axes[0].legend(fontsize=8); fig.suptitle('K-means : on alterne « assigner au centroïde le plus proche » et « recalculer les centroïdes »', fontsize=10); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (marketing) :** segmentation de clientèle selon récence, fréquence et montant d'achat (RFM).
- **Cas d'usage 2 :** regroupement de pays selon leurs indicateurs de commerce extérieur pour définir des marchés cibles.
- **Données et prérequis :** variables numériques standardisées, groupes approximativement sphériques ; choix de $K$ par coude ou silhouette.
- **Peu approprié :** groupes de formes allongées ou imbriquées (anneaux, croissants) ou nombreuses valeurs aberrantes.
- **Métriques :** silhouette, inertie, indice de Davies-Bouldin ; stabilité entre initialisations ; ARI si des étiquettes de référence existent.

```python
from sklearn.cluster import KMeans
modele=KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(X_clu)
print('Silhouette:',silhouette_score(X_clu,modele.labels_))
plt.scatter(X_clu[:,0],X_clu[:,1],c=modele.labels_,s=12);plt.title('K-means');plt.show()
```

Les étiquettes `etiquettes_reference` ne sont pas utilisées dans cette cellule.

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** classification ascendante hiérarchique (*agglomerative hierarchical clustering*). **Famille :** clustering. **Objectif :** construire une hiérarchie de groupes emboîtés puis la couper au niveau voulu.
- **Intuition :** chaque point commence seul dans son groupe ; on fusionne progressivement les deux groupes les plus proches jusqu'à n'en avoir plus qu'un.
- **Étapes :** matrice de distances → fusion répétée des deux groupes minimisant un critère de liaison → construction du dendrogramme → coupure pour obtenir $K$ groupes.
- **Critère :** avec la liaison de Ward (défaut de scikit-learn), on fusionne les groupes $A$ et $B$ qui augmentent le moins la variance interne : $\Delta(A,B)=\dfrac{|A||B|}{|A|+|B|}\|\mu_A-\mu_B\|^2$, où $|A|$ est la taille du groupe $A$ et $\mu_A$ son centre.
- **Paramètres / hyperparamètres :** il n'y a pas de paramètres appris au sens classique ; le nombre de groupes `n_clusters`, le critère de liaison (`linkage`) et la distance sont des **hyperparamètres**.
- **Avantages :** (1) le dendrogramme montre plusieurs niveaux de regroupement ; (2) pas besoin d'initialisation aléatoire.
- **Limites :** (1) coût quadratique en mémoire et en temps, donc peu adapté aux gros jeux ; (2) une fusion faite ne peut plus être défaite.

### Schéma explicatif

**Comment lire ce schéma :** Chaque feuille en bas est une observation ; chaque trait horizontal représente une fusion, à une hauteur égale à la dissimilarité des groupes réunis. La ligne rouge en tirets coupe l'arbre et définit ici 3 groupes (couleurs différentes).

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 18
from scipy.cluster.hierarchy import linkage, dendrogram
g = np.random.default_rng(9)
X = np.vstack([g.normal([0, 0], .4, (5, 2)), g.normal([3, 3], .4, (5, 2)), g.normal([0, 4], .4, (4, 2))])
Z = linkage(X, method='ward')
fig, ax = plt.subplots(figsize=(7, 4))
dendrogram(Z, ax=ax, color_threshold=3.0)
ax.axhline(3.0, c='red', ls='--'); ax.text(ax.get_xlim()[1], 3.3, 'coupure → 3 groupes', color='red', fontsize=9, ha='right')
ax.set_xlabel('indices des observations'); ax.set_ylabel('distance de fusion (critère de Ward)')
ax.set_title('Classification hiérarchique : la hauteur d\'une fusion mesure la dissimilarité des groupes réunis', fontsize=9); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (commerce international) :** regrouper des pays ou des produits aux profils d'échanges semblables et visualiser les niveaux de proximité.
- **Cas d'usage 2 :** classer des entreprises d'un secteur selon leurs ratios financiers.
- **Données et prérequis :** variables numériques standardisées, échantillon de taille modérée (quelques milliers de points au plus).
- **Peu approprié :** très grands volumes ou données avec beaucoup de bruit (le chaînage de groupes peut déformer la hiérarchie).
- **Métriques :** silhouette, corrélation cophénétique, stabilité du découpage ; interprétation du dendrogramme.

```python
from sklearn.cluster import AgglomerativeClustering
etiquettes=AgglomerativeClustering(n_clusters=3).fit_predict(X_clu)
print('Silhouette:',silhouette_score(X_clu,etiquettes))
```

- **Sortie :** le score de silhouette des étiquettes obtenues.
- **Ce qu'on peut conclure :** si le score est proche de celui de K-means, les deux méthodes trouvent la même structure sur ces amas bien séparés. **Ce qu'on ne peut pas conclure :** la méthode n'a pas de graine et `AgglomerativeClustering` ne trace pas le dendrogramme (il faut utiliser SciPy) ; la silhouette ne dit rien du « bon » nombre de groupes sans comparer plusieurs valeurs de $K$.

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** DBSCAN (*Density-Based Spatial Clustering of Applications with Noise*). **Famille :** clustering (et détection de bruit). **Objectif :** trouver des groupes de forme quelconque d'après la densité locale et isoler le bruit.
- **Intuition :** un groupe est une zone dense de points reliés entre eux ; les points isolés dans les zones peu denses sont étiquetés comme bruit.
- **Étapes :** pour chaque point, comptage des voisins dans un rayon $\varepsilon$ → points « coeurs » s'ils ont au moins `min_samples` voisins → connexion des points coeurs voisins en groupes → rattachement des points frontière, le reste est du bruit (étiquette −1).
- **Critère :** un point $x$ est coeur si $|\{x'\,:\,\|x-x'\|\le\varepsilon\}|\ge \texttt{min\_samples}$, où $\varepsilon$ est le rayon de voisinage et `min_samples` le seuil de densité.
- **Paramètres / hyperparamètres :** les groupes obtenus ne sont pas des paramètres appris ; $\varepsilon$ (`eps`) et `min_samples` sont les **hyperparamètres** déterminants.
- **Avantages :** (1) détecte des formes arbitraires sans fixer le nombre de groupes ; (2) identifie explicitement le bruit.
- **Limites :** (1) très sensible à $\varepsilon$ et à `min_samples`, surtout si les densités des groupes diffèrent ; (2) performances dégradées en grande dimension.

### Schéma explicatif

**Comment lire ce schéma :** Les points pleins sont des points coeurs, les cercles vides des points frontière et les croix noires du bruit. Le cercle rouge en tirets montre le rayon $\varepsilon$ autour d'un point : on compte les voisins qu'il contient. Deux croissants de lune sont séparés, ce que K-means ne saurait pas faire.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 19
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
X, _ = make_moons(250, noise=.07, random_state=0); X = np.vstack([X, [[1.8, 1.3], [-1.2, 1.2], [0.5, -1.0]]])
eps = .15; m = DBSCAN(eps=eps, min_samples=5).fit(X)
coeur = np.zeros(len(X), bool); coeur[m.core_sample_indices_] = True
fig, ax = plt.subplots(figsize=(7.5, 4.2))
ax.scatter(*X[m.labels_ == -1].T, c='k', marker='x', s=50)
for k in set(m.labels_) - {-1}:
    col = plt.cm.tab10(k)
    ax.scatter(*X[(m.labels_ == k) & coeur].T, s=22, color=col); ax.scatter(*X[(m.labels_ == k) & ~coeur].T, s=26, facecolors='white', edgecolors=col)
c0 = X[m.core_sample_indices_[10]]
ax.add_patch(Circle(c0, eps, fill=False, ec='red', ls='--')); ax.annotate(r'rayon $\varepsilon$', xy=c0+[eps*.7, eps*.7], xytext=c0+[.35, .55], color='red', fontsize=9, arrowprops=dict(arrowstyle='->', color='red'))
from matplotlib.lines import Line2D
ax.legend(handles=[Line2D([], [], marker='o', ls='', c='gray', label='point coeur (≥ min_samples voisins)'), Line2D([], [], marker='o', ls='', mfc='white', mec='gray', label='point frontière'), Line2D([], [], marker='x', ls='', c='k', label='bruit (-1)')], fontsize=8, loc='lower left')
ax.set_aspect('equal'); ax.set_title('DBSCAN : groupes = zones denses reliées ; les points isolés sont du bruit', fontsize=9); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance) :** détecter des groupes de transactions denses et isoler les opérations atypiques (bruit) dans un espace de caractéristiques.
- **Cas d'usage 2 :** regrouper des points de livraison selon leurs coordonnées géographiques pour définir des zones de tournée.
- **Données et prérequis :** variables numériques standardisées ; une distance significative ; réglage de $\varepsilon$ (par exemple via le graphe des distances au $k$-ième voisin).
- **Peu approprié :** groupes de densités très différentes ou espace de grande dimension.
- **Métriques :** silhouette (hors bruit), proportion de bruit, stabilité face à de petites variations de $\varepsilon$ ; ARI si des étiquettes existent.

```python
from sklearn.cluster import DBSCAN
etiquettes=DBSCAN(eps=.8,min_samples=5).fit_predict(X_clu)
print('Groupes:',set(etiquettes),'Bruit:',(etiquettes==-1).sum())
```

### Analyse de l'expérience Python

- **Entrées :** `X_clu` (240 points 2D). **Hyperparamètres :** `eps=0.8`, `min_samples=5`.
- **Sortie :** l'ensemble des étiquettes trouvées (−1 = bruit) et le nombre de points de bruit.
- **Ce qu'on peut conclure :** combien de groupes DBSCAN détecte avec ce rayon et combien de points il juge isolés. **Ce qu'on ne peut pas conclure :** que ce nombre de groupes est « le bon » : il dépend directement de `eps` (qui n'a pas été réglé et dont l'échelle n'a pas été vérifiée) ; le code n'affiche pas de silhouette, donc aucune qualité n'est mesurée. Une valeur de `eps` différente donnerait un autre nombre de groupes et de points de bruit.
- **Résultat observé lors de notre exécution** (Python 3.13, scikit-learn 1.9.1, graine 42) : 3 groupes (étiquettes 0, 1, 2) et 8 points de bruit. Ces 8 points sont étiquetés « bruit » alors que le jeu ne contient que 3 amas : avec `eps=0.8`, des points en bordure d'amas sont jugés trop isolés. *Vérifiez ces valeurs sur votre propre exécution : elles peuvent légèrement varier selon les versions des bibliothèques.*

## 20. Mélanges gaussiens

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** modèle de mélange gaussien (*Gaussian Mixture Model*, GMM). **Famille :** clustering probabiliste. **Objectif :** modéliser les données comme un mélange de lois normales et en déduire des groupes « souples ».
- **Intuition :** chaque groupe est une loi normale (centre, forme, taille) ; chaque point appartient à tous les groupes avec une certaine probabilité.
- **Étapes :** initialisation → algorithme EM : étape E (calcul des probabilités d'appartenance) puis étape M (mise à jour des moyennes, covariances et poids) → répétition jusqu'à convergence.
- **Critère :** $p(x)=\sum_{k=1}^{K}\pi_k\,\mathcal N(x;\mu_k,\Sigma_k)$, où $\pi_k$ est le poids du composant $k$ ($\sum_k\pi_k=1$), $\mu_k$ sa moyenne et $\Sigma_k$ sa matrice de covariance ; on maximise la vraisemblance des données.
- **Paramètres / hyperparamètres :** $\pi_k,\mu_k,\Sigma_k$ sont les **paramètres** ; `n_components` ($K$), `covariance_type` et l'initialisation sont des **hyperparamètres**.
- **Avantages :** (1) groupes de formes elliptiques variées et probabilités d'appartenance ; (2) modèle génératif qui permet un choix de $K$ par BIC/AIC.
- **Limites :** (1) dépend de l'initialisation et peut converger vers un optimum local ; (2) suppose des composantes gaussiennes et devient instable en grande dimension.

### Schéma explicatif

**Comment lire ce schéma :** Chaque couple d'ellipses noires représente un composant gaussien à 1 et 2 écarts-types ; la croix rouge est sa moyenne. Les ellipses sont inclinées : le modèle capte la corrélation entre variables, ce que K-means (groupes sphériques) ne fait pas.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 20
from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture
X, _ = make_blobs(n_samples=300, centers=3, random_state=2); X = X @ np.array([[1, .6], [0, .8]])
m = GaussianMixture(3, random_state=0).fit(X)
fig, ax = plt.subplots(figsize=(6.5, 4.6))
ax.scatter(X[:, 0], X[:, 1], c=m.predict(X), cmap='viridis', s=12, alpha=.6)
for mu, cov in zip(m.means_, m.covariances_):
    val, vec = np.linalg.eigh(cov); ang = np.degrees(np.arctan2(vec[1, 1], vec[0, 1]))
    for n in (1, 2): ax.add_patch(Ellipse(mu, 2*n*np.sqrt(val[1]), 2*n*np.sqrt(val[0]), angle=ang, fill=False, ec='k', lw=1.2))
    ax.scatter(*mu, marker='X', s=100, c='red', ec='k')
ax.set_xlabel('variable 1'); ax.set_ylabel('variable 2'); ax.set_title(r'Mélange gaussien : chaque composante $\mathcal{N}(\mu_k,\Sigma_k)$ est tracée à 1 et 2 écarts-types', fontsize=9); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (marketing) :** segmentation souple de clients dont les profils se chevauchent, avec une probabilité d'appartenance par segment.
- **Cas d'usage 2 :** modélisation de la distribution des rendements financiers par un mélange de régimes (calme / turbulent).
- **Données et prérequis :** variables numériques, composantes à peu près gaussiennes, assez d'observations par composante.
- **Peu approprié :** formes de groupes très non gaussiennes (croissants) ou dimension élevée avec peu d'observations.
- **Métriques :** BIC, AIC, log-vraisemblance, silhouette ; stabilité entre initialisations.

```python
from sklearn.mixture import GaussianMixture
modele=GaussianMixture(n_components=3,random_state=GRAINE).fit(X_clu)
print('Poids:',modele.weights_,'Silhouette:',silhouette_score(X_clu,modele.predict(X_clu)))
```

### Analyse de l'expérience Python

- **Entrées :** `X_clu` (240 points 2D). **Hyperparamètres :** `n_components=3`, `random_state=42` ; type de covariance par défaut (`'full'`).
- **Sorties :** les poids des composantes (`weights_`, somme égale à 1) et la silhouette calculée sur les étiquettes dures (`predict`).
- **Ce qu'on peut conclure :** la proportion estimée de chaque groupe et la séparation des étiquettes obtenues. **Ce qu'on ne peut pas conclure :** l'incertitude d'affectation (les probabilités `predict_proba` ne sont pas utilisées, la silhouette n'utilise que le groupe le plus probable) ; que $K=3$ est optimal (BIC non comparé) ; des résultats réels, les données étant synthétiques et gaussiennes, donc favorables à ce modèle.
- **Résultat observé lors de notre exécution** (Python 3.13, scikit-learn 1.9.1, graine 42) : poids égaux (≈ 1/3 chacun) et silhouette ≈ 0,869, identique à K-means : cohérent avec trois amas de même taille. *Vérifiez ces valeurs sur votre propre exécution : elles peuvent légèrement varier selon les versions des bibliothèques.*

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** analyse en composantes principales (*Principal Component Analysis*, ACP / PCA). **Famille :** réduction de dimension (non supervisée). **Objectif :** résumer des données par quelques axes qui conservent le maximum de variance.
- **Intuition :** on cherche les directions le long desquelles les données varient le plus et on projette dessus ; la première composante capte le plus de variance, la suivante le plus possible de ce qui reste, orthogonalement.
- **Étapes :** centrage (et standardisation) → matrice de covariance → vecteurs propres triés par valeur propre décroissante → projection sur les $k$ premiers.
- **Critère :** $\max_{\|w\|=1}\ \mathrm{Var}(Xw)$ ; les composantes sont les vecteurs propres de $\Sigma$ et la part de variance expliquée par la composante $j$ vaut $\lambda_j/\sum_{l}\lambda_l$, où $\Sigma$ est la matrice de covariance et $\lambda_j$ la valeur propre associée.
- **Paramètres / hyperparamètres :** axes et valeurs propres = **paramètres** calculés ; le nombre de composantes `n_components` est l'**hyperparamètre**.
- **Avantages :** (1) déterministe, rapide, bien interprétable (poids des variables) ; (2) utile pour visualiser, débruiter ou décorréler.
- **Limites :** (1) linéaire : elle rate les structures courbes ; (2) sensible à l'échelle des variables et aux valeurs extrêmes.

### Schéma explicatif

**Comment lire ce schéma :** Le nuage est allongé selon une diagonale ; la flèche rouge (CP1) suit cette direction de variance maximale, la flèche verte (CP2) est perpendiculaire. Les segments gris sont les projections des points sur CP1. La légende donne la part de variance de chaque composante.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 21
from sklearn.decomposition import PCA
g = np.random.default_rng(10); X = g.normal(size=(120, 2)) @ np.array([[2.0, 1.2], [0, .6]]); X -= X.mean(0)
p = PCA(2).fit(X); u = p.components_[0]
proj = (X @ u)[:, None]*u
fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(*X.T, s=14, c='#2563eb', label='observations centrées')
for i in range(0, len(X), 4): ax.plot([X[i, 0], proj[i, 0]], [X[i, 1], proj[i, 1]], c='#d1d5db', lw=.7)
for k, col in zip(range(2), ['#dc2626', '#16a34a']):
    v = p.components_[k]*2*np.sqrt(p.explained_variance_[k]); ax.annotate('', xy=v, xytext=(0, 0), arrowprops=dict(arrowstyle='->', lw=2.2, color=col))
    ax.text(v[0]+.15, v[1]-.1 if k == 0 else v[1]+.2, f'CP{k+1} ({p.explained_variance_ratio_[k]*100:.0f} % de la variance)', color=col, fontsize=9, ha='left' if k == 0 else 'right')
ax.set_aspect('equal'); ax.set_xlim(-7, 9); ax.set_xlabel('variable 1'); ax.set_ylabel('variable 2'); ax.set_title('ACP : CP1 maximise la variance des projections (segments gris)', fontsize=9); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (finance) :** résumer des dizaines de taux d'intérêt de maturités différentes en 2 ou 3 facteurs (niveau, pente, courbure).
- **Cas d'usage 2 :** construire un indice composite à partir de nombreux indicateurs économiques de pays, ou visualiser un portefeuille de clients en 2D.
- **Données et prérequis :** variables numériques, standardisées si les unités diffèrent ; relations plutôt linéaires.
- **Peu approprié :** structures non linéaires (spirales, variétés courbes) ou variables catégorielles brutes.
- **Métriques :** variance expliquée (cumulée), éboulis, erreur de reconstruction.

```python
from sklearn.decomposition import PCA
X_vin=StandardScaler().fit_transform(load_wine().data)
modele=PCA(n_components=2).fit(X_vin)
projection=modele.transform(X_vin)
print('Variance expliquée:',modele.explained_variance_ratio_)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.show()
```

### Analyse de l'expérience Python

- **Entrées :** le jeu `load_wine` (vrai jeu de chimie des vins, 178 observations × 13 variables), standardisé par `StandardScaler`. **Hyperparamètre :** `n_components=2`.
- **Sorties :** la variance expliquée par chacune des deux composantes et un nuage de points projeté (non coloré).
- **Ce qu'on peut conclure :** la part d'information (variance) conservée par les 2 axes et la forme générale du nuage. **Ce qu'on ne peut pas conclure :** à quoi correspondent les axes (les poids des variables ne sont pas affichés) ; que les groupes visibles correspondent aux cépages (le nuage n'est pas coloré par classe, et la variance expliquée ne mesure pas la séparation des classes). Ici l'ACP est ajustée sur tout le jeu, sans découpage entraînement/test : acceptable pour de l'exploration mais pas pour évaluer un modèle.
- **Résultat observé lors de notre exécution** (Python 3.13, scikit-learn 1.9.1, graine 42) : variance expliquée ≈ 0,362 pour la 1re composante et ≈ 0,192 pour la 2e, soit environ 55 % de la variance totale : une projection 2D perd donc près de 45 % de l'information. *Vérifiez ces valeurs sur votre propre exécution : elles peuvent légèrement varier selon les versions des bibliothèques.*

## 22. t-SNE

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** t-SNE (*t-distributed Stochastic Neighbor Embedding*). **Famille :** réduction de dimension non linéaire, destinée à la visualisation. **Objectif :** représenter en 2D ou 3D les données de grande dimension en conservant les voisinages locaux.
- **Intuition :** les points proches dans l'espace d'origine doivent rester proches sur la carte ; on déplace les points de la carte jusqu'à ce que leurs similarités ressemblent à celles d'origine.
- **Étapes :** similarités $p_{ij}$ dans l'espace d'origine (noyau gaussien calibré par la perplexité) → initialisation de la carte → similarités $q_{ij}$ (loi de Student) → descente de gradient sur la divergence de Kullback-Leibler.
- **Critère :** $\min_{Y}\ \mathrm{KL}(P\|Q)=\sum_{i\ne j}p_{ij}\log\frac{p_{ij}}{q_{ij}}$, où $p_{ij}$ et $q_{ij}$ sont les similarités entre les points $i$ et $j$ dans l'espace d'origine et sur la carte $Y$.
- **Paramètres / hyperparamètres :** les coordonnées de la carte sont les **paramètres** optimisés ; `perplexity`, `learning_rate`, nombre d'itérations et initialisation sont des **hyperparamètres**.
- **Avantages :** (1) très bon pour révéler des amas locaux ; (2) cartes lisibles pour l'exploration.
- **Limites :** (1) les distances entre amas et leurs tailles n'ont pas de sens quantitatif, et la carte change avec la perplexité et la graine ; (2) pas de projection directe de nouveaux points et coût élevé sur gros jeux.

### Schéma explicatif

**Comment lire ce schéma :** Diagramme de flux de gauche à droite : on mesure les similarités d'origine ($p_{ij}$), on crée des points 2D avec leurs propres similarités ($q_{ij}$), puis on les déplace (flèche rouge) pour réduire l'écart entre les deux ensembles de similarités.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 22
fig, ax = canevas(11, 3.6, (0, 12), (0, 4.4))
boite(ax, 1.5, 2.2, 'Données en\nhaute dimension\n$x_i\\in\\mathbb{R}^p$', 2.4, 1.3, '#e5e7eb')
boite(ax, 4.6, 2.2, 'Similarités de\nvoisinage $p_{ij}$\n(noyau gaussien,\nperplexité)', 2.6, 1.3)
boite(ax, 7.8, 2.2, 'Points $y_i$ en 2D,\nsimilarités $q_{ij}$\n(loi de Student, 1 d.d.l.)', 2.8, 1.3, '#fef3c7')
boite(ax, 10.7, 2.2, 'Carte 2D', 1.6, 1.0, '#dcfce7')
fleche(ax, 2.8, 2.2, 3.3, 2.2); fleche(ax, 5.95, 2.2, 6.4, 2.2); fleche(ax, 9.2, 2.2, 9.9, 2.2)
ax.annotate('', xy=(5.5, 1.2), xytext=(7.8, 1.2), arrowprops=dict(arrowstyle='<->', lw=1.2, color='#b91c1c'))
ax.text(6.65, .6, r'on déplace les $y_i$ pour minimiser $KL(P\|Q)=\sum_{i\neq j}p_{ij}\log\frac{p_{ij}}{q_{ij}}$', ha='center', fontsize=9, color='#b91c1c')
ax.set_title('t-SNE : conserver les voisinages locaux, pas les distances globales', fontsize=10); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (marketing) :** explorer visuellement des profils clients décrits par de nombreuses variables pour repérer des segments possibles.
- **Cas d'usage 2 :** visualiser des représentations vectorielles de produits ou de documents.
- **Données et prérequis :** variables numériques standardisées ; quelques milliers de points sont confortables ; tester plusieurs perplexités.
- **Peu approprié :** comme étape de prétraitement avant un modèle (la carte n'est pas stable ni généralisable) ou pour mesurer des distances entre groupes.
- **Métriques :** conservation des voisinages (*trustworthiness*, continuité), divergence KL finale ; l'évaluation visuelle seule est trompeuse.

```python
from sklearn.manifold import TSNE
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=TSNE(n_components=2,perplexity=20,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('t-SNE');plt.show()
```

### Analyse de l'expérience Python

- **Entrées :** `load_wine` standardisé (178 × 13). **Hyperparamètres :** `n_components=2`, `perplexity=20`, `random_state=42`.
- **Sortie :** un nuage de points 2D (non coloré, aucune métrique affichée).
- **Ce qu'on peut conclure :** qualitativement, si des amas locaux apparaissent. **Ce qu'on ne peut pas conclure :** quoi que ce soit de quantitatif : les axes n'ont pas de signification, les distances entre amas ne sont pas interprétables, et rien ne montre que la structure vue est réelle plutôt qu'un artefact du réglage de la perplexité. Pour évaluer, il faudrait une mesure de conservation des voisinages (ex. `trustworthiness`) et la comparer à l'ACP.

## 23. UMAP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** UMAP (*Uniform Manifold Approximation and Projection*). **Famille :** réduction de dimension non linéaire. **Objectif :** projeter des données en basse dimension en conservant structure locale et, mieux que t-SNE, une partie de la structure globale.
- **Intuition :** on construit un graphe de voisinage pondéré des données, puis on cherche une disposition en 2D dont le graphe est le plus ressemblant.
- **Étapes :** recherche des $k$ plus proches voisins → graphe flou pondéré → initialisation spectrale → optimisation par descente de gradient d'une entropie croisée entre graphe d'origine et graphe de la carte.
- **Critère :** on minimise $\sum_{(i,j)}\big[w_{ij}\log\frac{w_{ij}}{v_{ij}}+(1-w_{ij})\log\frac{1-w_{ij}}{1-v_{ij}}\big]$, où $w_{ij}$ et $v_{ij}$ sont les poids d'arête entre $i$ et $j$ dans l'espace d'origine et sur la carte.
- **Paramètres / hyperparamètres :** coordonnées de la carte = **paramètres** optimisés ; `n_neighbors`, `min_dist`, `metric`, `n_components` = **hyperparamètres**.
- **Avantages :** (1) rapide sur de grands jeux et peut projeter de nouveaux points (`transform`) ; (2) compromis plus équilibré entre structure locale et globale.
- **Limites :** (1) résultat sensible aux hyperparamètres et à la graine, avec une interprétation quantitative des distances limitée ; (2) bibliothèque externe `umap-learn`.

## 24. Isolation Forest

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** forêt d'isolation (*Isolation Forest*). **Famille :** détection d'anomalies (non supervisée). **Objectif :** repérer les observations rares et atypiques.
- **Intuition :** un point atypique est isolé du reste par très peu de coupures aléatoires, alors qu'un point normal, noyé dans la masse, en demande beaucoup.
- **Étapes :** construction de nombreux arbres avec variables et seuils tirés au hasard → pour chaque point, longueur moyenne du chemin jusqu'à son isolement → conversion en score → seuil fixé par la contamination.
- **Critère :** $s(x,n)=2^{-E[h(x)]/c(n)}$, où $h(x)$ est la profondeur à laquelle $x$ est isolé dans un arbre, $E[h(x)]$ sa moyenne sur la forêt, $n$ la taille de l'échantillon et $c(n)$ une constante de normalisation ; un score proche de 1 signale une anomalie.
- **Paramètres / hyperparamètres :** arbres et seuil = **paramètres** ; `n_estimators`, `max_samples`, `contamination` (proportion attendue d'anomalies) = **hyperparamètres**.
- **Avantages :** (1) rapide, adapté aux gros jeux, sans hypothèse de distribution ; (2) fonctionne sans étiquettes.
- **Limites :** (1) le seuil dépend de `contamination`, souvent inconnue ; (2) moins efficace pour des anomalies cachées dans des sous-espaces ou des anomalies locales.

### Schéma explicatif

**Comment lire ce schéma :** Panneau de gauche : chaque point est coloré par son score d'anomalie ; les deux points éloignés du nuage ont les scores les plus élevés. Panneau de droite : quelques coupures aléatoires (traits bleus) ; un point éloigné est séparé du reste après très peu de coupures.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

- **Métriques :** sans étiquettes, qualité des alertes (revue d'un échantillon par un expert, stabilité du classement) ; avec étiquettes fiables, précision, rappel, PR-AUC.

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** bandit manchot à plusieurs bras (*multi-armed bandit*) ; l'exemple du notebook illustre le cas le plus simple, l'**estimation par tirage uniforme** des bras. **Famille :** apprentissage par renforcement (cas sans état). **Objectif :** maximiser le gain cumulé en choisissant entre plusieurs actions dont les gains sont inconnus.
- **Intuition :** il faut tester chaque bras pour connaître sa valeur, mais chaque essai sur un mauvais bras coûte un gain ; c'est le dilemme exploration / exploitation.
- **Étapes :** choix d'un bras → observation d'une récompense aléatoire → mise à jour de la moyenne estimée du bras → répétition.
- **Critère :** estimation $\hat\mu_a=\frac{1}{n_a}\sum_{t:\,a_t=a}r_t$ et regret cumulé $R_T=T\mu^*-\sum_{t=1}^{T}\mu_{a_t}$, où $a$ désigne un bras, $n_a$ le nombre de tirages du bras $a$, $r_t$ la récompense au tour $t$, $\mu^*$ la meilleure valeur moyenne possible et $T$ le nombre total de tours.
- **Paramètres / hyperparamètres :** les valeurs estimées $\hat\mu_a$ sont les **paramètres** appris ; la règle de choix et le nombre de tours sont les **hyperparamètres** (ici : rotation des bras `t % 3`).
- **Avantages :** (1) cadre simple et rigoureux pour décider sous incertitude ; (2) estimations sans biais avec assez d'essais.
- **Limites :** (1) le tirage uniforme n'exploite jamais l'information acquise (regret linéaire) ; (2) ignore le contexte (états, variables explicatives) et suppose des gains stables.

### Schéma explicatif

**Comment lire ce schéma :** À gauche : la boucle agent → bras → gain (0 ou 1). À droite : après 100 tirages de chaque bras, les barres bleues (moyennes estimées) s'approchent des points rouges (vraies probabilités, cachées à l'agent en pratique).

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 25
g = np.random.default_rng(42); vraies = np.array([.2, .5, .7]); somme = np.zeros(3); n = np.zeros(3)
for t in range(300):
    b = t % 3; somme[b] += g.random() < vraies[b]; n[b] += 1
fig, axes = plt.subplots(1, 2, figsize=(11, 3.8), gridspec_kw={'width_ratios': [1.3, 1]})
ax = axes[0]; ax.set_xlim(0, 10); ax.set_ylim(0, 4); ax.axis('off')
boite(ax, 1.5, 2, 'Agent', 1.8, 1.0, '#e5e7eb')
for k, y in enumerate([3.2, 2.0, 0.8]): boite(ax, 6, y, f'Bras {k+1}\n(proba inconnue)', 2.6, .9); fleche(ax, 2.5, 2, 4.7, y); fleche(ax, 7.3, y, 8.5, 2)
boite(ax, 9.3, 2, 'gain\n0 ou 1', 1.5, 1.0, '#dcfce7'); ax.text(2.2, 3.6, 'tire un bras', fontsize=8); ax.set_title('boucle : tirer un bras → observer un gain', fontsize=9)
ax = axes[1]; ax.bar(['bras 1', 'bras 2', 'bras 3'], somme/n, color='#93c5fd', label='moyenne estimée'); ax.scatter(range(3), vraies, c='red', zorder=3, label='vraie proba (cachée)')
ax.set_ylim(0, 1); ax.legend(fontsize=8); ax.set_title('tirage uniforme : 100 essais par bras', fontsize=9)
fig.suptitle('Bandit manchot : estimer la valeur de chaque bras par des essais répétés', fontsize=10); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (marketing) :** test de plusieurs versions d'une bannière ou d'un email pour trouver celle qui a le meilleur taux de clic.
- **Cas d'usage 2 :** choix dynamique d'un fournisseur ou d'un transporteur selon le taux de livraison à l'heure (logistique).
- **Données et prérequis :** une récompense mesurable après chaque choix ; peu d'actions ; environnement à peu près stable.
- **Peu approprié :** quand la décision dépend d'un contexte (état du client) et d'effets durables : il faut alors un bandit contextuel ou un vrai apprentissage par renforcement.
- **Métriques :** regret cumulé, récompense cumulée, taux de choix du meilleur bras, variation entre graines.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3)
for t in range(500):
    bras=t%3
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Valeurs estimées:',somme/comptes)
```

### Analyse de l'expérience Python

- **Entrées :** aucune donnée : un générateur aléatoire (`default_rng(42)`) simule 3 bras de probabilités de gain 0,2, 0,5 et 0,7. **Choix des bras :** `t % 3`, c'est-à-dire en alternance parfaite, **sans stratégie** (exploration uniforme).
- **Sortie :** la moyenne de gain estimée pour chaque bras après 500 tours (environ 167 tirages par bras).
- **Ce qu'on peut conclure :** avec assez de tirages, les estimations se rapprochent des vraies probabilités et permettent de classer les bras. **Ce qu'on ne peut pas conclure :** que l'algorithme « optimise » quelque chose : une rotation fixe donne un regret élevé (le mauvais bras est tiré autant que le bon). Il s'agit d'une simulation, pas de données réelles ; l'exemple suivant (epsilon-greedy) ajoute la prise de décision.
- **Résultat observé lors de notre exécution** (Python 3.13, scikit-learn 1.9.1, graine 42) : valeurs estimées ≈ 0,18 ; 0,54 ; 0,72, proches des vraies probabilités (0,2 ; 0,5 ; 0,7). *Vérifiez ces valeurs sur votre propre exécution : elles peuvent légèrement varier selon les versions des bibliothèques.*

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** stratégie epsilon-greedy (*ε-greedy*). **Famille :** apprentissage par renforcement (bandits). **Objectif :** gagner un maximum en équilibrant exploration (essayer) et exploitation (jouer le meilleur bras connu).
- **Intuition :** la plupart du temps on joue le bras qui paraît le meilleur ; de temps en temps, avec probabilité $\varepsilon$, on joue un bras au hasard pour continuer à apprendre.
- **Étapes :** tirage d'un nombre $u$ → si $u<\varepsilon$, bras aléatoire, sinon bras de plus grande valeur estimée → observation du gain → mise à jour de l'estimation.
- **Critère :** $a_t=\begin{cases}\text{bras aléatoire} & \text{avec probabilité }\varepsilon\ \arg\max_a\hat\mu_a & \text{avec probabilité }1-\varepsilon\end{cases}$, où $\hat\mu_a$ est la valeur estimée du bras $a$ et $\varepsilon\in[0,1]$ le taux d'exploration.
- **Paramètres / hyperparamètres :** les $\hat\mu_a$ sont les **paramètres** appris ; $\varepsilon$ (ici 0,1) et le nombre de tours sont les **hyperparamètres**. Le code initialise l'estimation de chaque bras à 1 via $(S+1)/(n+1)$, une initialisation « optimiste » qui pousse à essayer chaque bras au début.
- **Avantages :** (1) très simple à implémenter ; (2) garantit que tous les bras continuent d'être explorés.
- **Limites :** (1) une exploration constante coûte du regret même quand le meilleur bras est connu ; (2) l'exploration est aveugle (elle ne cible pas les bras incertains).

### Schéma explicatif

**Comment lire ce schéma :** À gauche : l'arbre de décision à chaque tour (explorer ou exploiter). À droite : regret cumulé de trois valeurs de $\varepsilon$ sur une simulation unique. Cette courbe illustre le mécanisme, mais elle dépend de la graine et de l'initialisation optimiste : elle ne prouve pas qu'une valeur de $\varepsilon$ est supérieure.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 26
fig, axes = plt.subplots(1, 2, figsize=(11.5, 3.8), gridspec_kw={'width_ratios': [1.4, 1]})
ax = axes[0]; ax.set_xlim(0, 10); ax.set_ylim(0, 4.4); ax.axis('off')
boite(ax, 1.4, 2.2, 'tirer\n$u\\sim U(0,1)$', 2.0, 1.0, '#e5e7eb'); boite(ax, 4.6, 3.4, '$u<\\varepsilon$ ?\nOUI : explorer\n(bras au hasard)', 2.9, 1.1, '#fef3c7'); boite(ax, 4.6, 1.0, 'NON : exploiter\n(bras de meilleure\nvaleur estimée)', 2.9, 1.1, '#dcfce7')
boite(ax, 8.6, 2.2, 'gain observé →\nmise à jour de\nla moyenne du bras', 2.7, 1.2)
fleche(ax, 2.4, 2.4, 3.2, 3.2); fleche(ax, 2.4, 2.0, 3.2, 1.2); fleche(ax, 6.1, 3.2, 7.4, 2.5); fleche(ax, 6.1, 1.2, 7.4, 1.9)
ax.set_title('règle de décision à chaque tour', fontsize=9)
g = np.random.default_rng(42); vraies = np.array([.2, .5, .7])
for eps, col in [(0.0, '#dc2626'), (.1, '#2563eb'), (.5, '#16a34a')]:
    somme = np.zeros(3); n = np.zeros(3); regret = []; r = 0
    for t in range(1000):
        b = int(g.integers(3)) if g.random() < eps else int(np.argmax((somme+1)/(n+1)))
        somme[b] += g.random() < vraies[b]; n[b] += 1; r += vraies.max()-vraies[b]; regret.append(r)
    axes[1].plot(regret, c=col, label=rf'$\varepsilon$ = {eps}')
axes[1].set_xlabel('tour'); axes[1].set_ylabel('regret cumulé espéré'); axes[1].legend(fontsize=8); axes[1].set_title('une seule simulation (graine 42) : illustration, pas une preuve', fontsize=8.5)
fig.suptitle('Epsilon-greedy : arbitrer entre explorer et exploiter', fontsize=10); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (marketing / e-commerce) :** choisir dynamiquement quelle offre promotionnelle afficher pour maximiser les achats.
- **Cas d'usage 2 :** allocation de ressources publicitaires entre plusieurs canaux dont le rendement est inconnu.
- **Données et prérequis :** retour d'information rapide après chaque action ; récompenses stables dans le temps ; nombre d'actions limité.
- **Peu approprié :** environnements qui changent vite (il faut une estimation à mémoire courte) ou décisions à fort coût d'erreur (une exploration aléatoire est inacceptable, par exemple en santé).
- **Métriques :** récompense cumulée, regret cumulé, fréquence de sélection du meilleur bras, variation entre graines.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

### Analyse de l'expérience Python

- **Entrées :** simulation de 3 bras (0,2 ; 0,5 ; 0,7), `default_rng(42)`. **Hyperparamètres :** `epsilon=0.1`, 1000 tours, estimation $(somme+1)/(comptes+1)$.
- **Sorties :** le nombre de choix de chaque bras et les valeurs estimées.
- **Ce qu'on peut conclure :** si le bras 3 (le meilleur) concentre l'essentiel des choix, la stratégie a bien exploité l'information. **Ce qu'on ne peut pas conclure :** que $\varepsilon=0{,}1$ est optimal ; que le résultat est reproductible sur d'autres graines (une seule exécution, aucune moyenne) ; l'estimation d'un bras très peu joué reste imprécise. Il manque aussi une mesure explicite du regret ou de la récompense cumulée.
- **Résultat observé lors de notre exécution** (Python 3.13, scikit-learn 1.9.1, graine 42) : choix ≈ [25, 49, 926] et valeurs estimées ≈ [0,20 ; 0,67 ; 0,69] : le bras 3 (le meilleur) est joué dans environ 93 % des tours. L'estimation du bras 2 (≈ 0,67 pour une vraie valeur de 0,5) est nettement biaisée à cause de ses 49 tirages seulement. *Vérifiez ces valeurs sur votre propre exécution : elles peuvent légèrement varier selon les versions des bibliothèques.*

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Q-learning (*Q-learning*). **Famille :** apprentissage par renforcement (sans modèle, hors politique, tabulaire). **Objectif :** apprendre une politique d'actions qui maximise la récompense cumulée.
- **Intuition :** l'agent apprend une table $Q(s,a)$ qui estime le gain futur total si l'on fait l'action $a$ dans l'état $s$ puis que l'on joue au mieux ensuite.
- **Étapes :** observer l'état $s$ → choisir une action (stratégie ε-greedy) → recevoir $r$ et le nouvel état $s'$ → mettre à jour $Q(s,a)$ → répéter sur de nombreux épisodes.
- **Critère :** $Q(s,a)\leftarrow Q(s,a)+\alpha\,[\,r+\gamma\max_{a'}Q(s',a')-Q(s,a)\,]$, où $\alpha$ est le taux d'apprentissage, $\gamma$ le facteur d'actualisation (poids du futur), $r$ la récompense, $s'$ l'état suivant et $a'$ une action possible dans $s'$.
- **Paramètres / hyperparamètres :** la table $Q$ contient les **paramètres** appris ; $\alpha$ (0,3), $\gamma$ (0,95), $\varepsilon$ (qui décroît de 1 à 0,05), nombre d'épisodes (4000) sont des **hyperparamètres**.
- **Avantages :** (1) simple, converge vers la politique optimale dans un cas tabulaire avec assez d'exploration ; (2) hors politique : apprend la politique optimale tout en explorant.
- **Limites :** (1) inutilisable si le nombre d'états est très grand ou continu (la table explose) ; (2) sensible aux hyperparamètres et à la conception de la récompense.

### Schéma explicatif

**Comment lire ce schéma :** L'agent (à gauche) envoie une action à l'environnement (à droite), qui renvoie un nouvel état et une récompense. La formule en bas met à jour la table $Q$ ; son terme $\max_{a'}$ prend le meilleur coup possible à l'état suivant, quelle que soit l'action réellement jouée ensuite.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 27
fig, ax = canevas(11, 4.2, (0, 12), (0, 5.4))
boite(ax, 2.2, 3.9, 'Agent\n(table $Q(s,a)$)', 2.8, 1.2, '#dbeafe'); boite(ax, 9.2, 3.9, 'Environnement\n(FrozenLake)', 2.8, 1.2, '#dcfce7')
fleche(ax, 3.7, 4.2, 7.7, 4.2, 'action $a_t$ (ε-greedy)', .15); fleche(ax, 7.7, 3.6, 3.7, 3.6, 'état $s_{t+1}$, récompense $r_{t+1}$', -.4)
ax.text(6, 1.5, r'$Q(s_t,a_t)\leftarrow Q(s_t,a_t)+\alpha\,[\,r_{t+1}+\gamma\,\max_{a^{\prime}}Q(s_{t+1},a^{\prime})-Q(s_t,a_t)\,]$', ha='center', fontsize=11)
ax.text(6, .7, r'la cible utilise le MEILLEUR coup possible en $s_{t+1}$ (apprentissage « hors politique »)', ha='center', fontsize=9, color='#b91c1c')
ax.set_title('Q-learning : boucle agent–environnement et règle de mise à jour', fontsize=10); plt.show()
```

- **Données et prérequis :** un environnement avec états et actions discrets, un simulateur permettant beaucoup d'interactions, une récompense bien définie.
- **Peu approprié :** espaces d'états très grands ou continus (préférer DQN ou des approximateurs de fonction) ; situations où l'on ne peut pas expérimenter librement (finance réelle, santé).
- **Métriques :** récompense cumulée par épisode, taux de réussite, vitesse de convergence, variation entre graines.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** SARSA (*State–Action–Reward–State–Action*). **Famille :** apprentissage par renforcement (sans modèle, sur la politique, tabulaire). **Objectif :** apprendre une politique en évaluant la valeur des actions réellement jouées.
- **Intuition :** comme Q-learning, mais la mise à jour utilise l'action suivante **effectivement choisie** (donc exploration incluse), ce qui rend l'agent plus prudent dans les environnements risqués.
- **Étapes :** choisir $a$ dans $s$ → observer $r$, $s'$ → choisir $a'$ dans $s'$ avec la même politique → mettre à jour $Q(s,a)$ avec $a'$ → continuer avec $(s',a')$.
- **Critère :** $Q(s,a)\leftarrow Q(s,a)+\alpha\,[\,r+\gamma\,Q(s',a')-Q(s,a)\,]$, où $\alpha$ est le taux d'apprentissage, $\gamma$ le facteur d'actualisation et $a'$ l'action effectivement choisie dans $s'$.
- **Paramètres / hyperparamètres :** la table $Q$ = **paramètres** ; $\alpha$, $\gamma$, $\varepsilon$ et le nombre d'épisodes = **hyperparamètres**.
- **Avantages :** (1) tient compte de l'exploration, d'où des politiques plus sûres dans les environnements à risque ; (2) simple et stable.
- **Limites :** (1) apprend la politique suivie (avec exploration) plutôt que la politique optimale pure, sauf si $\varepsilon\to 0$ ; (2) mêmes limites tabulaires que Q-learning.

### Schéma explicatif

**Comment lire ce schéma :** Même boucle agent–environnement que pour Q-learning ; seule change la formule de mise à jour : elle utilise $Q(s_{t+1},a_{t+1})$, la valeur de l'action réellement choisie ensuite, au lieu du maximum.

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 28
fig, ax = canevas(11, 4.2, (0, 12), (0, 5.4))
boite(ax, 2.2, 3.9, 'Agent\n(table $Q(s,a)$)', 2.8, 1.2, '#dbeafe'); boite(ax, 9.2, 3.9, 'Environnement\n(FrozenLake)', 2.8, 1.2, '#dcfce7')
fleche(ax, 3.7, 4.2, 7.7, 4.2, 'action $a_t$ (ε-greedy)', .15); fleche(ax, 7.7, 3.6, 3.7, 3.6, 'état $s_{t+1}$, récompense $r_{t+1}$', -.4)
ax.text(6, 1.5, r'$Q(s_t,a_t)\leftarrow Q(s_t,a_t)+\alpha\,[\,r_{t+1}+\gamma\,Q(s_{t+1},a_{t+1})-Q(s_t,a_t)\,]$', ha='center', fontsize=11)
ax.text(6, .7, r"la cible utilise l'action $a_{t+1}$ RÉELLEMENT choisie (ε-greedy comprise) : apprentissage « sur la politique »", ha='center', fontsize=9, color='#b91c1c')
ax.set_title('SARSA : même boucle que Q-learning, mais la cible dépend de l\'action suivante effectivement jouée', fontsize=10); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (logistique) :** apprendre une politique de conduite ou de gestion d'entrepôt où les actions exploratoires peuvent être coûteuses et doivent être prises en compte.
- **Cas d'usage 2 :** robotique ou navigation dans un environnement comportant des zones dangereuses (exemple classique : « Cliff Walking »).
- **Données et prérequis :** environnement discret avec simulateur, récompenses définies ; décroissance contrôlée de ε.
- **Peu approprié :** grands espaces d'états (préférer approximation de fonction) ; objectif de politique optimale pure quand l'exploration est sans conséquence.
- **Métriques :** récompense cumulée, taux de réussite, sécurité du chemin appris, variation entre graines.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

### Analyse de l'expérience Python

- **Entrées :** `FrozenLake-v1` déterministe (16 états, 4 actions), table `Q` de taille 16 × 4. **Hyperparamètres :** $\alpha=0{,}3$, $\gamma=0{,}95$, 4000 épisodes, ε de 1 à 0,05.
- **Sortie :** la politique gloutonne `Q.argmax(axis=1)`.
- **Ce qu'on peut conclure :** quelles actions SARSA privilégie dans les états visités ; on peut comparer cette liste à celle de Q-learning (exemple 27). **Ce qu'on ne peut pas conclure :** que SARSA est « plus prudent » ou « moins bon » : sur une grille déterministe sans zone de danger asymétrique, les deux méthodes peuvent donner des politiques identiques ou proches. Aucune récompense cumulée n'est mesurée, une seule graine est utilisée, et les états non visités affichent l'action 0 par défaut.
- **Résultat observé lors de notre exécution** (Python 3.13, scikit-learn 1.9.1, graine 42) : politique `[1 0 3 0 1 0 1 0 2 1 1 0 0 2 2 0]`. Elle diffère de celle de Q-learning uniquement aux états 1 et 2, qui ne sont pas sur le chemin principal ; depuis l'état 0, la trajectoire est la même que pour Q-learning. *Vérifiez ces valeurs sur votre propre exécution : elles peuvent légèrement varier selon les versions des bibliothèques.*

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** réseau Q profond (*Deep Q-Network*, DQN). **Famille :** apprentissage par renforcement profond. **Objectif :** étendre Q-learning aux espaces d'états de grande taille ou continus en approchant $Q(s,a)$ par un réseau de neurones.
- **Intuition :** au lieu d'une table, un réseau de neurones prédit les valeurs de toutes les actions à partir de l'état ; deux astuces stabilisent l'apprentissage : une mémoire de rejeu et un réseau cible.
- **Étapes :** l'agent joue et stocke ses transitions dans la mémoire → on tire un mini-lot de transitions → on calcule la cible de Bellman avec le réseau cible → on met à jour le réseau en ligne par descente de gradient → on recopie périodiquement le réseau en ligne dans le réseau cible.
- **Critère :** cible $y=r+\gamma(1-d)\max_{a'}Q_{\theta^-}(s',a')$ et perte $\big(y-Q_\theta(s,a)\big)^2$, où $\theta$ sont les poids du réseau en ligne, $\theta^-$ ceux du réseau cible, $d$ vaut 1 si l'épisode se termine et 0 sinon.
- **Paramètres / hyperparamètres :** les poids $\theta$ sont les **paramètres** ; $\gamma$, taux d'apprentissage, taille de la mémoire, taille de lot, fréquence de recopie et $\varepsilon$ sont des **hyperparamètres**.
- **Avantages :** (1) gère des états de grande dimension (images, capteurs) ; (2) mémoire de rejeu et réseau cible rendent l'apprentissage plus stable.
- **Limites :** (1) très gourmand en interactions et sensible aux hyperparamètres ; (2) actions discrètes uniquement et possible surestimation des valeurs.

### Schéma explicatif

**Comment lire ce schéma :** L'environnement fournit des transitions qui sont stockées dans la mémoire de rejeu ; le réseau en ligne apprend sur des mini-lots tirés de cette mémoire ; la cible est calculée avec le réseau cible, une copie retardée du réseau en ligne (flèche en tirets : recopie périodique).

*Schéma original produit avec Matplotlib (aucune figure publiée n'est reprise).*

```python
# Schéma explicatif — algorithme 29
fig, ax = canevas(11.5, 4.6, (0, 12), (0, 6))
boite(ax, 1.6, 3.0, 'Environnement', 2.4, 1.0, '#dcfce7'); boite(ax, 5.0, 5.0, 'Mémoire de rejeu\n$(s,a,r,s^{\\prime},\\text{fin})$', 2.9, 1.1, '#fef3c7')
boite(ax, 5.0, 3.0, 'Réseau en ligne\n$Q_\\theta(s,a)$', 2.6, 1.1, '#dbeafe'); boite(ax, 9.7, 3.0, 'Réseau cible\n$Q_{\\theta^-}(s^{\\prime},a^{\\prime})$', 2.6, 1.1, '#e5e7eb')
boite(ax, 7.3, 0.9, r'cible $y=r+\gamma(1-\mathrm{fin})\max_{a^{\prime}}Q_{\theta^-}(s^{\prime},a^{\prime})$', 5.4, .9, '#fee2e2')
fleche(ax, 2.8, 3.3, 3.6, 4.6, 'transitions'); fleche(ax, 5.0, 4.4, 5.0, 3.6, 'mini-lot'); fleche(ax, 9.7, 2.4, 8.6, 1.3); fleche(ax, 5.0, 2.4, 6.0, 1.3, 'perte $(y-Q_\\theta)^2$')
ax.annotate('', xy=(6.4, 3.0), xytext=(8.4, 3.0), arrowprops=dict(arrowstyle='<-', ls='--', color='gray')); ax.text(7.4, 3.2, 'copie périodique de $\\theta$', ha='center', fontsize=8, color='gray'); fleche(ax, 3.7, 3.0, 2.8, 3.0, 'action')
ax.set_title('DQN : réseau en ligne, réseau cible et mémoire de rejeu', fontsize=10); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 1 (logistique / gestion) :** pilotage dynamique de tarifs ou de stocks dans un simulateur à nombreuses variables d'état.
- **Cas d'usage 2 :** jeux vidéo (Atari) ou contrôle de systèmes simulés à partir d'images ou de capteurs.
- **Données et prérequis :** simulateur rapide, très grand nombre d'interactions, bibliothèque de réseaux de neurones (PyTorch, TensorFlow) ; actions discrètes.
- **Peu approprié :** problèmes où l'on ne peut pas simuler ou expérimenter, petits problèmes discrets (une table suffit) ou actions continues (préférer d'autres algorithmes).
- **Métriques :** récompense cumulée moyenne, taux de réussite, variation entre graines, courbe d'apprentissage.

Découpage supervisé : 60 % entraînement / 20 % validation / 20 % test (`random_state=42`, stratifié pour la classification).

```python
# Outils communs : découpage 60/20/20 et sélection d'hyperparamètres sur la validation
def split3(X, y, stratify=False):
    s = y if stratify else None
    X_tmp, X_te, y_tmp, y_te = train_test_split(X, y, test_size=.2, random_state=SEED, stratify=s)
    s2 = y_tmp if stratify else None
    X_tr, X_va, y_tr, y_va = train_test_split(X_tmp, y_tmp, test_size=.25, random_state=SEED, stratify=s2)
    return X_tr, X_va, X_te, y_tr, y_va, y_te

def selectionner(fabrique, grille, X_tr, y_tr, X_va, y_va, score, maximiser):
    meilleur = None
    for params in ParameterGrid(grille):
        m = fabrique(**params).fit(X_tr, y_tr)
        s = score(y_va, m.predict(X_va))
        if meilleur is None or (s > meilleur[0] if maximiser else s < meilleur[0]):
            meilleur = (s, params)
    return meilleur[1], meilleur[0]

def rmse(y, p):
    return float(np.sqrt(mean_squared_error(y, p)))
```

## Q7/Q8 — Clustering (Wine) : K-means, hiérarchique, DBSCAN, mélanges gaussiens
Les variables sont standardisées. Les étiquettes (cultivars) **ne servent jamais à l'apprentissage ni au choix de k** : seulement à une évaluation **externe** (ARI, NMI). Critères internes : silhouette (plus haut = mieux) et indice de Davies-Bouldin (plus bas = mieux). Stabilité : ARI moyen entre la partition complète et celles obtenues sur 10 sous-échantillons de 80 %.

```python
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.mixture import GaussianMixture

Xw = StandardScaler().fit_transform(df_wine.drop(columns='target').to_numpy()); yw = df_wine['target'].to_numpy()

# Choix de k par critère interne (silhouette), sans utiliser les étiquettes
sil_k = {k: silhouette_score(Xw, KMeans(k, n_init=10, random_state=SEED).fit_predict(Xw)) for k in range(2, 9)}
k_opt = max(sil_k, key=sil_k.get)
print('Silhouette K-means selon k:', {k: round(v, 3) for k, v in sil_k.items()}, '→ k retenu =', k_opt)

# DBSCAN : choix (eps, min_samples) par silhouette hors bruit (≥ 2 groupes, ≤ 30 % de bruit)
meilleur_db, sc_db = (3.0, 5), -1
for eps in np.arange(1.5, 4.51, .25):
    for ms in (3, 5, 10):
        lab = DBSCAN(eps=eps, min_samples=ms).fit_predict(Xw); mask = lab != -1
        if len(set(lab[mask])) >= 2 and mask.mean() >= .7:
            s = silhouette_score(Xw[mask], lab[mask])
            if s > sc_db: sc_db, meilleur_db = s, (float(eps), ms)
print('DBSCAN (eps, min_samples) retenus:', meilleur_db)

clu_modeles = {
    'K-means (réf.)': lambda: KMeans(k_opt, n_init=10, random_state=SEED),
    'Hiérarchique (Ward)': lambda: AgglomerativeClustering(n_clusters=k_opt, linkage='ward'),
    'DBSCAN': lambda: DBSCAN(eps=meilleur_db[0], min_samples=meilleur_db[1]),
    'Mélange gaussien': lambda: GaussianMixture(n_components=k_opt, random_state=SEED),
}

def metriques_clust(lab):
    mask = lab != -1
    if len(set(lab[mask])) < 2:
        return np.nan, np.nan
    return silhouette_score(Xw[mask], lab[mask]), davies_bouldin_score(Xw[mask], lab[mask])

def stabilite(fab, complet, n=10):
    rng = np.random.default_rng(SEED); aris = []
    for _ in range(n):
        idx = rng.choice(len(Xw), size=int(.8 * len(Xw)), replace=False)
        aris.append(adjusted_rand_score(complet[idx], fab().fit_predict(Xw[idx])))
    return np.mean(aris), np.std(aris)

lignes, labels_clu = [], {}
for nom, fab in clu_modeles.items():
    t0 = time.time(); lab = fab().fit_predict(Xw); t = time.time() - t0; labels_clu[nom] = lab
    sil, dbi = metriques_clust(lab); st = stabilite(fab, lab)
    lignes.append({'Méthode': nom, 'Groupes': len(set(lab) - {-1}), 'Bruit (%)': 100 * (lab == -1).mean(), 'Silhouette': sil,
                   'Davies-Bouldin': dbi, 'ARI (externe)': adjusted_rand_score(yw, lab), 'NMI (externe)': normalized_mutual_info_score(yw, lab),
                   'Stabilité (ARI)': f"{st[0]:.3f} ± {st[1]:.3f}", 'Temps (s)': t})
res_clu = pd.DataFrame(lignes).set_index('Méthode')
display(res_clu.round(3))
print("Note : silhouette et Davies-Bouldin de DBSCAN sont calculés hors points de bruit ; ARI/NMI comptent le bruit comme un groupe.")

# Visualisation dans le plan ACP (projection seulement pour l'affichage)
from sklearn.decomposition import PCA
Z = PCA(2, random_state=SEED).fit_transform(Xw)
fig, ax = plt.subplots(1, 4, figsize=(16, 3.8), sharex=True, sharey=True)
for a, (nom, lab) in zip(ax, labels_clu.items()):
    a.scatter(Z[:, 0], Z[:, 1], c=lab, cmap='tab10', s=14); a.set_title(nom, fontsize=10); a.set_xlabel('CP1')
ax[0].set_ylabel('CP2'); fig.suptitle('Wine — partitions projetées sur les 2 premières composantes (couleur = groupe trouvé)', y=1.03)
plt.tight_layout(); plt.show()
```

## Q7/Q8 — Détection d'anomalies (Breast Cancer, protocole de détection de nouveauté)
**Protocole :** les modèles apprennent uniquement sur des tumeurs **bénignes** (70 % d'entre elles) ; le test contient les 30 % de bénignes restantes + des tumeurs **malignes** tirées au hasard pour représenter ≈ 10 % du test (anomalies rares). Les étiquettes (diagnostic réel) servent **seulement à l'évaluation**. La répétition sur 10 tirages montre la variabilité. Référence : score z maximal par variable. Précision@k = rappel@k quand k = nombre réel d'anomalies du test.

```python
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.svm import OneClassSVM

ben, mal = np.flatnonzero(yb == 0), np.flatnonzero(yb == 1)

def score_z(Xa, Xt):
    mu, sd = Xa.mean(0), Xa.std(0) + 1e-9
    return np.abs((Xt - mu) / sd).max(1)

detecteurs = {
    'Score z max (réf.)': None,
    'Isolation Forest': lambda: make_pipeline(StandardScaler(), IsolationForest(n_estimators=200, random_state=SEED)),
    'LOF (novelty)': lambda: make_pipeline(StandardScaler(), LocalOutlierFactor(n_neighbors=20, novelty=True)),
    'One-Class SVM': lambda: make_pipeline(StandardScaler(), OneClassSVM(nu=.05, gamma='scale')),
}
res = {n: {'ROC-AUC': [], 'PR-AUC': [], 'P@k=R@k': []} for n in detecteurs}
for i in range(10):
    rng = np.random.default_rng(SEED + i); b = rng.permutation(ben); n_tr = int(.7 * len(b))
    tr, te_b = b[:n_tr], b[n_tr:]; te_m = rng.choice(mal, size=int(round(len(te_b) / 9)), replace=False)
    te = np.concatenate([te_b, te_m]); y_true = np.concatenate([np.zeros(len(te_b)), np.ones(len(te_m))]); k = len(te_m)
    for nom, fab in detecteurs.items():
        s = score_z(Xb[tr], Xb[te]) if fab is None else -fab().fit(Xb[tr]).decision_function(Xb[te])
        top = np.argsort(-s)[:k]
        res[nom]['ROC-AUC'].append(roc_auc_score(y_true, s)); res[nom]['PR-AUC'].append(average_precision_score(y_true, s))
        res[nom]['P@k=R@k'].append(y_true[top].mean())
res_ano = pd.DataFrame({n: {m: pm(v) for m, v in d.items()} for n, d in res.items()}).T
print(f"Test type : {len(te_b)} bénignes + {len(te_m)} malignes (prévalence des anomalies ≈ {100*len(te_m)/len(te):.0f} %). Un score aléatoire donnerait un PR-AUC ≈ {len(te_m)/len(te):.2f}.")
display(res_ano)
fig, ax = plt.subplots(figsize=(6.5, 3.8))
moy = pd.Series({n: np.mean(d['ROC-AUC']) for n, d in res.items()}); err = pd.Series({n: np.std(d['ROC-AUC']) for n, d in res.items()})
moy.plot.barh(xerr=err, ax=ax, color='#E15759'); ax.set_xlim(.5, 1); ax.set_xlabel('ROC-AUC (moyenne ± écart-type, 10 tirages)')
ax.set_title("Breast Cancer — détection d'anomalies"); plt.tight_layout(); plt.show()
```

## Q7/Q8 — Bandits manchots : **simulation** (pas un dataset réel)
Gymnasium ne fournit pas d'environnement de bandit dans ses environnements de base : on simule donc des bras de Bernoulli de probabilités connues `[0,1 ; 0,3 ; 0,5 ; 0,6 ; 0,7]`. **Ce sont des données simulées**, utilisées uniquement pour comparer des stratégies (20 graines, 2000 tirages). Référence : alternance uniforme. Métriques : récompense cumulée, regret cumulé (écart à la meilleure option), part de choix du meilleur bras.

```python
P_BRAS = np.array([.1, .3, .5, .6, .7])

def bandit(strategie, graine, T=2000, eps=.1):
    rng = np.random.default_rng(graine); K = len(P_BRAS); n = np.zeros(K); s = np.zeros(K)
    gains, regret, meilleur = np.zeros(T), np.zeros(T), np.zeros(T)
    for t in range(T):
        if strategie == 'Uniforme (réf.)':
            a = t % K
        elif n.min() == 0:
            a = int(np.argmin(n))
        elif strategie == 'ε-greedy (ε=0,1)':
            a = int(rng.integers(K)) if rng.random() < eps else int(np.argmax(s / n))
        else:  # UCB1
            a = int(np.argmax(s / n + np.sqrt(2 * np.log(t + 1) / n)))
        r = float(rng.random() < P_BRAS[a]); n[a] += 1; s[a] += r
        gains[t] = r; regret[t] = P_BRAS.max() - P_BRAS[a]; meilleur[t] = a == P_BRAS.argmax()
    return gains, np.cumsum(regret), meilleur

strats = ['Uniforme (réf.)', 'ε-greedy (ε=0,1)', 'UCB1']
lignes, courbes = [], {}
for st in strats:
    out = [bandit(st, SEED + g) for g in range(20)]
    G = np.array([o[0].sum() for o in out]); R = np.array([o[1] for o in out]); B = np.array([o[2].mean() for o in out])
    courbes[st] = R; lignes.append({'Stratégie': st, 'Récompense cumulée': pm(G, 1), 'Regret final': pm(R[:, -1], 1), 'Part du meilleur bras': pm(B)})
res_band = pd.DataFrame(lignes).set_index('Stratégie'); display(res_band)
fig, ax = plt.subplots(figsize=(6.5, 3.8))
for st, R in courbes.items():
    m, sd = R.mean(0), R.std(0); ax.plot(m, label=st); ax.fill_between(range(len(m)), m - sd, m + sd, alpha=.2)
ax.set_xlabel('Tirage'); ax.set_ylabel('Regret cumulé'); ax.set_title('Bandits (simulation) — regret cumulé, moyenne ± écart-type (20 graines)')
ax.legend(); plt.tight_layout(); plt.show()
```

## Q7/Q8 — Renforcement : Q-learning vs SARSA sur Gymnasium `FrozenLake-v1` (glissant)
**Simulation interactive, pas un dataset réel** : les trajectoires sont générées par l'agent dans l'environnement public Gymnasium. Variante `is_slippery=True` (stochastique, plus réaliste que la version déterministe du notebook). 5 graines × 5000 épisodes ; α=0,1, γ=0,99, ε décroissant de 1 à 0,05. Évaluation : politique **gloutonne** sur 1000 épisodes nouveaux (taux de réussite), récompense moyenne des 500 derniers épisodes d'entraînement, récompense cumulée. Référence : politique aléatoire. Le **DQN complet est facultatif et n'est pas réalisé ici** (le DQN du notebook reste conceptuel).

```python
import gymnasium as gym
print('gymnasium', gym.__version__)
EPISODES, ALPHA, GAMMA = 5000, .1, .99

def choix_glouton(q, rng):
    return int(rng.choice(np.flatnonzero(q == q.max())))

def entrainer(algo, graine):
    env = gym.make('FrozenLake-v1', is_slippery=True); rng = np.random.default_rng(graine)
    Q = np.zeros((env.observation_space.n, env.action_space.n)); rec = np.zeros(EPISODES)
    for ep in range(EPISODES):
        eps = max(.05, 1 - ep / (.6 * EPISODES))
        s, _ = env.reset(seed=graine) if ep == 0 else env.reset()
        a = int(rng.integers(4)) if rng.random() < eps else choix_glouton(Q[s], rng)
        fini = False
        while not fini:
            s2, r, term, trunc, _ = env.step(a)
            a2 = int(rng.integers(4)) if rng.random() < eps else choix_glouton(Q[s2], rng)
            suiv = Q[s2].max() if algo == 'Q-learning' else Q[s2, a2]
            Q[s, a] += ALPHA * (r + GAMMA * (0. if term else suiv) - Q[s, a])
            rec[ep] += r; s, a = s2, a2; fini = term or trunc
    env.close(); return Q, rec

def evaluer(politique, graine, n=1000):
    env = gym.make('FrozenLake-v1', is_slippery=True); succes = 0
    for ep in range(n):
        s, _ = env.reset(seed=graine) if ep == 0 else env.reset(); fini = False
        while not fini:
            s, r, term, trunc, _ = env.step(politique(s)); fini = term or trunc
        succes += r
    env.close(); return succes / n

lignes, courbes_rl = [], {}
rng_alea = np.random.default_rng(SEED)
alea = [evaluer(lambda s: int(rng_alea.integers(4)), SEED + g) for g in range(5)]
lignes.append({'Méthode': 'Politique aléatoire (réf.)', 'Réussite (politique gloutonne)': pm(alea), 'Récompense moy. 500 derniers ép.': '—', 'Récompense cumulée': '—'})
for algo in ['Q-learning', 'SARSA']:
    reussite, fin, cumul, courbes_rl[algo] = [], [], [], []
    for g in range(5):
        Q, rec = entrainer(algo, SEED + g)
        reussite.append(evaluer(lambda s: int(Q[s].argmax()), SEED + 100 + g)); fin.append(rec[-500:].mean()); cumul.append(rec.sum()); courbes_rl[algo].append(rec)
    lignes.append({'Méthode': algo, 'Réussite (politique gloutonne)': pm(reussite), 'Récompense moy. 500 derniers ép.': pm(fin), 'Récompense cumulée': pm(cumul, 0)})
res_rl = pd.DataFrame(lignes).set_index('Méthode'); display(res_rl)

fig, ax = plt.subplots(figsize=(6.5, 3.8))
for algo, L in courbes_rl.items():
    M = np.array([np.convolve(r, np.ones(200) / 200, mode='valid') for r in L]); m, sd = M.mean(0), M.std(0)
    ax.plot(m, label=algo); ax.fill_between(range(len(m)), m - sd, m + sd, alpha=.2)
ax.set_xlabel('Épisode'); ax.set_ylabel('Récompense moyenne (fenêtre 200)'); ax.set_title('FrozenLake glissant — courbes d\'apprentissage (5 graines)')
ax.legend(); plt.tight_layout(); plt.show()
```

## 5. Discussion

**Performance.** Sur ces petits jeux, les méthodes avancées ne battent pas nettement une référence bien réglée : régression linéaire en régression, régression logistique en classification. Les écarts de test (de 1 à 4 points de F1, ou de 2 unités de RMSE) sont du même ordre que les écarts-types de la validation croisée, donc non concluants. Le résultat à retenir est celui de l'arbre unique, systématiquement dernier, et la nécessité d'une baseline : le R² de la moyenne est proche de 0.

**Coût de calcul.** Tous les entraînements durent moins d'une demi-seconde sur ces jeux (le plus lent : AdaBoost, 0,30 s ; la forêt, 0,18 à 0,30 s ; t-SNE, 0,29 s). Ces durées ne permettent pas d'extrapoler à de gros volumes, où forêts, boosting, SVM et t-SNE coûtent bien plus que les modèles linéaires, K-means ou l'ACP.

**Interprétabilité.** Les modèles linéaires, l'arbre peu profond et l'ACP offrent des coefficients, des règles ou des axes lisibles. Forêts, boosting, SVM à noyau, t-SNE et UMAP sont plus opaques ; ici, ce surcoût d'opacité n'est pas compensé par un gain mesuré, sauf pour la visualisation (t-SNE).

**Limites.**

- Jeux petits (442, 569 et 178 observations) ; tests de 89 à 114 lignes : résultats bruités.
- Hyperparamètres choisis sur de petites grilles ; un réglage plus fin pourrait changer les classements.
- Protocole d'anomalies = détection de nouveauté sur un jeu médical, pas un cas de fraude réel ; PR-AUC dépendant de la prévalence choisie (10 %).
- Les écarts ne sont pas testés statistiquement ; les écarts-types servent de repère seulement.
- Bandits et FrozenLake sont des simulations ; le DQN complet n'a pas été réalisé.

## 6. Conclusion

Sur des données réelles de petite taille, une méthode de référence simple (régression linéaire, régression logistique, K-means, ACP) reste compétitive : les méthodes avancées apportent peu, et pas de façon démontrée. Elles se justifient surtout par un besoin précis : visualisation non linéaire (t-SNE), arbitrage exploration-exploitation (ε-greedy ou UCB1) ou détection de points atypiques (Isolation Forest). Pour aller plus loin, il faudrait des jeux plus volumineux, des grilles de réglage plus larges et des tests statistiques sur les écarts. Le choix d'un algorithme doit partir de la tâche, des données et du besoin d'interprétation, pas de la sophistication du modèle.
